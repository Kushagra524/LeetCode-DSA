```python
from typing import List

class Solution:
  def heightChecker(self , heights : List[int]) -> int:
    expected = sorted(heights)
    count = 0
    n = len(heights)

    for i in range(n):
      if heights[i] != expected[i]:
        count += 1

    return count

s = Solution()
print(s.heightChecker(heights = [5 , 1 , 2 , 3 , 4]))
```
