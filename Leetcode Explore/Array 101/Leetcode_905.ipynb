```python
from typing import List

class Solution:
  def sortParity(self , nums : List[int]) -> List[int]:
    n = len(nums)
    start = 0

    for i in range(n):
      if nums[i] % 2 == 0:
        nums[start] , nums[i] = nums[i] , nums[start]
        start = start + 1

    return nums

s = Solution()
print(s.sortParity(nums = [3 , 2 , 4 , 1 , 5 , 9 , 6]))
```
