```python
class ListNode:
  def __init__(self , val = 0 , next = None):
    self.val = val
    self.next = next

class Solution:
  def linkedListHelper(self , arr):
    head = ListNode(arr[0])
    curr = head

    for val in arr[1 : ]:
      curr.next = ListNode(val)
      curr = curr.next

    return head

  def hasCycle(self , head : ListNode | None) -> bool:
    slow = head
    fast = head

    while fast != None and fast.next != None:
      slow = slow.next
      fast = fast.next

      if slow == fast:
        return True

    return False

  def run(self , arr):
    head = self.linkedListHelper(arr)
    print(self.hasCycle(head))

s = Solution()
print(s.run(arr = ))
```
