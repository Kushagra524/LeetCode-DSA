```python
from typing import List

class Solution:
  def peakElement(self , nums : List[int]) -> int:
    n = len(nums)
    peak = n -1
    l = 0
    r = n - 1

    while l < r:
      mid = (l + r) // 2

      if nums[mid] < nums[mid + 1]:    # this means greater element already present at right side
        l = mid + 1

      else:
        peak = mid
        r = mid

    return peak

s = Solution()
```

```python
print(s.peakElement(nums = [0,10,5,2]))
```
