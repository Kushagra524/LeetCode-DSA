```python
from typing import List

class Solution:
    def findRelativeRanks(self, score: List[int]) -> List[str]:
      s1 = sorted(score , reverse = True)
      rank = {}

      for i in range(len(score)):
        if i == 0:
          rank[s1[i]] = "Gold Medal"
        elif i == 1:
          rank[s1[i]] = "Silver Medal"
        elif i == 2:
          rank[s1[i]] = "Third"
        else:
          rank[s1[i]] = str(i + 1)

      res = []
      for x in score:
        res.append(rank[x])

      return res

s = Solution()
print(s.findRelativeRanks(score = [10,3,8,9,4]))
```
