***Normal Approach*** - *but this approach takes so much time to execute*

```python
from typing import List

class Solution:
  def singleNumber(self , nums : List[int]) -> int:
    if len(nums) == 1:
      return nums[0]

    for i in nums:
      if nums.count(i) == 1:
        return i

s = Solution()
```

```python
print(s.singleNumber(nums = [2 , 2 , 1 , 1 , 1 ,3 , 4 , 4]))
```

***Optimized Approach*** - *using XOR*

1. x ^ x = 0
2. x ^ 0 = x

but this logic only works when the non-unique numbers occur in a pair of two , becoz only then they give cancel out each other and at last we get the unique number in answer.

image.png

```python
class OptmizedSol:
  def singleNum(self , nums : List[int]) -> int:
    ans = 0
    for num in nums:
      ans = ans ^ num
    return ans
```

```python
opt = OptmizedSol()
nums_list = [2, 2, 1, 1, 4]
print(opt.singleNum(nums_list))
```

***Another approach*** - *using frequency map*

```python
import collections

class Sol:
  def uniqueNum(self , nums : List[int]) -> int:
    freq_map = collections.Counter(nums)

    for i in freq_map:
      if freq_map[i] == 1:
        return i

sol = Sol()
```

```python
print(sol.uniqueNum(nums = [2 , 2 , 1 , 1 , 4 , 3 , 4 , 4]))
```

```python
print(sol.uniqueNum(nums = [4,1,2,1,2]))
```
