```python
from typing import List

class Solution:
  def isPrefixOfWord(self , sentence : str , searchWord : str) -> int:
    words = sentence.split()       # I used this function to seperate words out of the string & store it in an array so that traversal can be easy

    for word in words:
      if word.startswith(searchWord):    # startswith() finds out if a sub-word is a prefix of a given word
        return words.index(word) + 1   # + 1 becoz we are given a 1-indexed string

    return -1

s = Solution()
```

```python
print(s.isPrefixOfWord(sentence = "i love eating burger", searchWord = "burg"))
```
