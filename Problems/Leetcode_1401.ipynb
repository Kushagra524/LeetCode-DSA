```python
class Solution:
  def checkOverlap(self , radius : int , xCenter : int , yCenter : int , x1 : int , y1 : int , x2 : int , y2 : int) -> bool:
    Cx = xCenter
    Cy = yCenter
    Xmin = min(x1 , x2)
    Xmax = max(x1 , x2)
    Ymin = min(y1 , y2)
    Ymax = max(y1 , y2)

    Px = max(Xmin , min(Cx , Xmax))
    Py = max(Ymin , min(Cy , Ymax))

    d = (Cx - Px)**2 + (Cy - Py)**2

    if d <= radius**2:
      return True
    else:
      return False

s = Solution()
print(s.checkOverlap(radius = 1, xCenter = 0, yCenter = 0, x1 = 1, y1 = -1, x2 = 3, y2 = 1))
```
