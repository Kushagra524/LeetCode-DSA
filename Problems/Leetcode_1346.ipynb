```python
from typing import List

class Sol:
  def checkifExist(self , arr : List[int]) -> bool:
    seen = set()

    for i in arr:
      if 2 * i in seen:
        return True

      if i % 2 == 0 and i // 2 in seen:
        return True

      seen.add(i)

    return False

s = Sol()
print(s.checkifExist(arr = [10 , 2 , 5 , 3]))
print(s.checkifExist(arr = [10 , 2 , 9 , 3]))
```
