```python
from typing import List

class SOlution:
  def longestCommonPrefix(self , strs : List[str]) -> str:
    prefix = strs[0]   # I treat prefix as base word for comaprarison with other words

    for word in strs[1 : ]:   # I start travering from 2nd word till last word
      while not word.startswith(prefix):
        prefix = prefix[ : -1]   # if not satisfied the condition , I remove the last character of prefix

        if prefix == "":
          return ""
    return prefix

s = SOlution()
```

```python
print(s.longestCommonPrefix(strs = ["flower","flow","flight"]))
```

```python
print(s.longestCommonPrefix(strs = ["dog","racecar","car"]))
```
