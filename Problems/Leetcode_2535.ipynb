```python
from typing import List

class Solution:
  def differenceOfSum(self , nums : List[int]) -> int:
    element_sum = 0
    digit_sum = 0
    for i in nums:
      element_sum = element_sum + i

      while i > 0:
        last_digit = i % 10
        digit_sum = digit_sum + last_digit
        i = i // 10
    return abs(element_sum - digit_sum)


s = Solution()
```

```python
print(s.differenceOfSum(nums = [1 , 15 , 6 , 3]))
```

```python
print(s.differenceOfSum(nums = [1 , 12 , 31 , 42]))
```
