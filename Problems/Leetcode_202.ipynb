```python
class Solution:
  def detectHappyNum(self , n : int) -> bool:
    occured = set()

    while n != 1:
      if n in occured:
        return False

      occured.add(n)

      sum = 0
      while n > 0:
        sum = sum + ((n % 10) ** 2)
        n = n // 10

      n = sum

    return True

s = Solution()
```

```python
print(s.detectHappyNum(n = 9))
```

```python
print(s.detectHappyNum(n = 19))
```
