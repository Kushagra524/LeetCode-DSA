```python
class Solution:
  def xorOperation(self , n : int , start : int) -> int:
    nums = []
    for i in range(n):
      nums.append(start + 2 * i)

    ans = 0
    for num in nums:
      ans = ans ^ num

    return ans

s = Solution()
```

```python
print(s.xorOperation(n = 5 , start = 0))
```
