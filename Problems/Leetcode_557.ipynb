***First Approach*** - using manual logic

```python
class Solution:
    def reverseWords(self, s: str) -> str:
        res = []
        word = []

        for ch in s:
          if ch != " ":
            word.append(ch)
          else:
            while word:
              res.append(word.pop())
            res.append(" ")
        while word:
          res.append(word.pop())

        return "".join(res)

s = Solution()
```

```python
print(s.reverseWords(s = "Let's take LeetCode contest"))
```

***Second Approach***

```python
class Solution:
  def reverseWords(self , s : str) -> str:
    words = s.split()
    res = []

    for word in words:
      res.append(word[ : :  -1])

    r = " ".join(res)
    return r

s = Solution()
```

```python
print(s.reverseWords(s = "Let's take LeetCode contest"))
```
