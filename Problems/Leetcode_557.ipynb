```python
class Solution:
    def reverseWords(self, s: str) -> str:
        res = []
        word = []

        for ch in s:
          if ch != " ":
            word.append(ch)
          else:
            while word:
              res.append(word.pop())
            res.append(" ")
        while word:
          res.append(word.pop())

        return "".join(res)

s = Solution()
```

```python
print(s.reverseWords(s = "Let's take LeetCode contest"))
```
