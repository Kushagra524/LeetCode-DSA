```python
from typing import List

class Solution:
    def countNegatives(self, grid: list[list[int]]) -> int:
        count = 0

        for i in grid:
            for j in i:
                if j < 0:
                    count = count + 1

        return count
```

```python
s = Solution()
```

```python
print(s.countNegatives(grid = [[4,3,2,-1],[3,2,1,-1],[1,1,-1,-2],[-1,-1,-2,-3]]))
```
