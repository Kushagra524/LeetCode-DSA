```python
# intervals = [[1,2],[2,3],[3,4],[1,3]]

from typing import List

class Solution:
  def eraseOverlapIntervals(self , intervals : List[List[int]]) -> int:
    intervals.sort(key = lambda x : x[1])   # this means I wanna sort my intervals based on the index = 1 of each element.

    # so intervals = [ [1 , 2] , [1 , 3] , [2 , 3] , [3 , 4] ]
    n = len(intervals)
    prev = 0
    count = 1   # count = 1 because assume that 1st interval is considered surely
    for i in range(1 , n):
      if intervals[i][0] >= intervals[prev][1]:
        count = count + 1
        prev = i


    return n - count

s = Solution()
print(s.eraseOverlapIntervals(intervals = [[1,2],[2,3],[3,4],[1,3]]))
```

```python
from typing import List

class Solution:
    def eraseOverlapIntervals(self, intervals: List[List[int]]) -> int:
        if not intervals:
            return 0

        # Sort in-place by end time to achieve O(1) extra space complexity
        intervals.sort(key=lambda x: x[1])

        # Track the end time of the last active non-overlapping interval
        end_time = intervals[0][1]
        non_overlapping_count = 1

        # Pythonic iteration avoids repeated list indexing, which is faster
        for start, end in intervals[1:]:
            if start >= end_time:
                non_overlapping_count += 1
                end_time = end # Update with the current interval's end time

        return len(intervals) - non_overlapping_count

s = Solution()
print("Test 1:", s.eraseOverlapIntervals([[1,2],[2,3],[3,4],[1,3]]))
print("Test 2:", s.eraseOverlapIntervals([[1,100],[11,22],[1,11],[2,12]]))
```
