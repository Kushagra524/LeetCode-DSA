# ***Explanation***
This code replaces each number in a list with the **largest number to its right**, and puts **-1** at the very end.

### How it works:
1. It starts from the **right side (end)** of the list and moves backwards to the left.
2. It keeps track of the **maximum value** seen so far.
3. For each position, it updates the list element with that maximum value, then updates the maximum value if the current number is larger.

```python
from typing import List

class Solution:
  def replaceElements(self , arr : List[int]) -> List[int]:
    n = len(arr)
    max_right = -1

    for i in range(n - 1 , -1 , -1):
      curr = arr[i]
      arr[i] = max_right
      max_right = max(max_right , curr)

    return arr

s = Solution()
print(s.replaceElements(arr = [17 , 18 , 5 , 4 , 6 , 1]))
```
