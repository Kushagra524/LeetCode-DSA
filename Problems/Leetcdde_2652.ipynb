```python
from typing import List

class Solution:
  def findTheArrayConcVal(self , nums : List[int]) -> int:
    n = len(nums)
    l = 0
    r = n - 1
    ans = 0

    while l < r:
      base = int(str(nums[l]) + str(nums[r]))
      ans = ans + base
      l = l + 1
      r = r - 1

    if l == r:
      ans = ans + nums[l]

    return ans

s = Solution()
print(s.findTheArrayConcVal(nums = [5,12,13,84]))
```
