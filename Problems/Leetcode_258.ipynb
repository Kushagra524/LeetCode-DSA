***First Approach*** - Time Complexity = O(n log n)

```python
class Solution:
  def addDigits(self , num : int) -> int:
    if num < 10:
      return num

    while num >= 10:          # if my number is greater than 10 , only then it will split and its digits would be added to sum

      sum = 0

      while num > 0:
        last_digit = num % 10
        sum = sum + last_digit
        num = num // 10

      # main logic of this code is line below
      num = sum  # at last to continue the loop , we assign num = sum , to again go into loop and proceed further

    return sum

s = Solution()
```

```python
print(s.addDigits(num = 38))
```

```python
print(s.addDigits(num = 3))
```

***Best approach*** - *Mathematical way*

1. Numbers divisible by 9: If a number is a multiple of 9 (e.g., 9, 18, 27, etc.), its digital root is always 9. If you calculate (num - 1) % 9, for num = 9, (9-1)%9 = 8%9 = 8. Adding 1 gives 9. For num = 18, (18-1)%9 = 17%9 = 8. Adding 1 gives 9. This holds true.

2. Numbers not divisible by 9: If a number is not a multiple of 9, its digital root is simply num % 9. For example, the digital root of 38 is 2 (3+8=11, 1+1=2). Using the formula: (38-1)%9 + 1 = 37%9 + 1 = 1 + 1 = 2. For 3, (3-1)%9 + 1 = 2%9 + 1 = 2 + 1 = 3.

3. The reason (num - 1) % 9 is used is to handle the case of numbers exactly divisible by 9. Without subtracting 1, num % 9 would return 0 for multiples of 9, which is incorrect as their digital root is 9.

```python
class OptmizedSolution:
  def addDigits(self , num : int) -> int:
    return ((num - 1) % 9) + 1

opt = OptmizedSolution()
```

```python
print(opt.addDigits(num = 98731))
# 9 + 8 + 7 + 3 + 1 = 28 ===> 2 + 8 = 10 ===> 1 + 0 = 1
```
