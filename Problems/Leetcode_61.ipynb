```python
class ListNode:
  def __init__(self , val = 0 , next = None):
    self.val = val
    self.next = next


class Solution:
  def linkedListHelper(self , arr):
    if not arr:
      return None

    head = ListNode(arr[0])
    curr = head

    for data in arr[1 : ]:
      curr.next = ListNode(data)
      curr = curr.next

    return head

  def rightRotate(self , head : ListNode | None , k : int) -> ListNode | None:
    if head is None or head.next is None:
      return head

    l = 0
    lastnode = head
    # I use lastnode.next != None so that 'lastnode' stops exactly on the very last node.
    # If I used lastnode != None, lastnode would become None, losing our reference to the last node.
    while lastnode.next != None:
      l = l + 1
      lastnode = lastnode.next
    l = l + 1   # +1 more becoz lastnode is made to reach 1 node before last node intentionally

    if k > l:
      k = k % l
    if k == l or k == 0:
      return head

    curr = head
    for i in range(l - k - 1):
      curr = curr.next

    lastnode.next = head      # if above lastnode was forces to traverse till actuall 'last node' while I was calculating length , then , now at this line it would be an error , becoz lastnode.next throws error ofc.
    head = curr.next
    curr.next = None

    return head

  def run(self , arr , k):
    head = self.linkedListHelper(arr)
    newHead = self.rightRotate(head , k)

    curr = newHead
    while curr != None:
      print(curr.val , end = " ")
      curr = curr.next

s = Solution()
res = s.run(arr = [1,2,3,4,5,6,7] , k = 3)
```
