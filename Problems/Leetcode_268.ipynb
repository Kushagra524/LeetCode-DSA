```python
from typing import List

class Solution:
  def missingNum(self , nums : list[int]) -> int:
    n = len(nums)   # to find the range of numbers

    sum_of_elements_of_nums = sum(nums)

    sum_of_n_numbers = (n * (n + 1)) // 2

    return (sum_of_n_numbers - sum_of_elements_of_nums)

s = Solution()
```

```python
print(s.missingNum(nums = [9,6,4,2,3,5,7,0,1]))
```
