```python
from typing import List

class Solution:
  def searchInsert(self , nums : List[int] , target : int) -> int:
    left = 0
    right = len(nums) - 1

    while left <= right:
      mid = (left + right) // 2

      if nums[mid] == target:
        return mid

      elif nums[mid] < target:
        left = mid + 1

      else:
        right = mid - 1

    return left

s = Solution()
```

```python
print(s.searchInsert(nums = [1 , 3 , 5 , 7] , target = 5))
```

```python
print(s.searchInsert(nums = [1 , 3 , 5 , 7] , target = 2))
```
