```python
class Solution:
  def countGoodString(self , s : str) -> int:
    answer = 0

    for i in range(0 , len(s) - 2):
      if s[i] != s[i+1] and s[i] != s[i + 2] and s[i + 1] != s[i + 2]:
        answer = answer + 1

    return answer

s = Solution()
```

```python
print(s.countGoodString(s = "xyzzaz"))
```

```python
print(s.countGoodString(s = "aababcabc"))
```
