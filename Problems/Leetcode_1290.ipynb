```python
from typing import List, Optional

class ListNode:
  def __init__(self, val=0, next=None):
    self.val = val
    self.next = next

class Solution:
  def linkedlistHelper(self, arr: List[int]) -> Optional[ListNode]:
    if not arr:
      return None
    head = ListNode(arr[0])
    curr = head
    for val in arr[1:]:
      curr.next = ListNode(val)
      curr = curr.next
    return head

  def getDecimalValue(self, head: Optional[ListNode]) -> int:
    curr = head
    binary = ""
    while curr is not None:
      binary += str(curr.val)
      curr = curr.next
    return int(binary, 2)

s = Solution()
head = s.linkedlistHelper([1, 0, 1])
print(s.getDecimalValue(head))
```
