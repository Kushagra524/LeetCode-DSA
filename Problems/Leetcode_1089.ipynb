````python
from typing import List

class Sol:
  def duplicateZeros(self , arr : List[int]) -> List[int]:
    n = len(arr)
    zero = arr.count(0)
    i = n - 1
    j = n - 1 + zero

    while i < j:
      if j < n:
        arr[j] = arr[i]
      j = j - 1

      if arr[i] == 0:
        if j < n:
          arr[j] = 0
        j = j - 1

      i = i - 1

    return arr

s = Sol()
print(s.duplicateZeros(arr = [1,0,2,3,0,4,5,0]))
````
