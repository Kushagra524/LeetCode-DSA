*First approach - frequency map* - ***taking too much time***

```python
import collections
from typing import List

class Solution:
  def containsDuplicate(self , nums : List[int]) -> bool:
    freq = collections.Counter(nums)

    for num in freq:
      if freq[num] >= 2:
        return True
    else:
      return False

s = Solution()
```

```python
print(s.containsDuplicate(nums = [1,1,1,3,3,4,3,2,4,2]))
```

```python
print(s.containsDuplicate(nums = [1 , 2 , 3 , 4]))
```

***Another approach - using set*** - *takes less time*

```python
from typing import List

class optmizedAproach:
  def hasDuplicate(self , nums : List[int]) -> bool:
    return len(nums) != len(set(nums))

opt = optmizedAproach()
```

```python
print(opt.hasDuplicate(nums = [1,1,1,3,3,4,3,2,4,2]))
```

```python
print(opt.hasDuplicate(nums = [1 , 2 , 3 , 4]))
```

### Third Approach - Sorting the array

This method involves sorting the input list first. Once sorted, duplicate elements will be adjacent to each other. We can then iterate through the sorted list and check if any element is equal to its next element.

```python
from typing import List

class SortingApproach:
  def containsDuplicate(self, nums: List[int]) -> bool:
    nums.sort()  # Sort the array in-place
    for i in range(len(nums) - 1):
      if nums[i] == nums[i+1]:
        return True
    return False

s_approach = SortingApproach()
```

```python
print(s_approach.containsDuplicate(nums = [1,1,1,3,3,4,3,2,4,2]))
```

```python
print(s_approach.containsDuplicate(nums = [1 , 2 , 3 , 4]))
```

```python
print(s_approach.containsDuplicate(nums = []))
```

```python
print(s_approach.containsDuplicate(nums = [1]))
```
