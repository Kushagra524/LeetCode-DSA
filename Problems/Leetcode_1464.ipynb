***First Approach***

```python
from typing import List

class Solution:
  def maxProduct(self , nums : List[int]) -> int:
    nums.sort()
    n = len(nums)
    max_product = 1

    for i in range(n - 1):
      product = (nums[i] - 1) * (nums[i + 1] - 1)
      max_product = max(max_product , product)

    return max_product

s = Solution()
```

```python
print(s.maxProduct(nums = [3 , 4 , 2 , 5]))
```

***Second Approach***

```python
class Sol:
  def maxProd(self , nums : List[int]) -> int:
    nums.sort()

    return (nums[-1] - 1) * (nums[-2] - 1)

s = Sol()
```

```python
print(s.maxProd(nums = [3 , 4 , 2 , 5]))
```
