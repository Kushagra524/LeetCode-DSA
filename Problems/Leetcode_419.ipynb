***Brute-Force Approach***

```python
from typing import List

class Solution:
    def thirdMax(self, nums: List[int]) -> int:
        s = list(set(nums))
        s.sort()

        for i in range(len(s)):
            if len(s) < 3:
                return max(s)
            elif len(s) == 3:
                return min(s)

        return s[-3]

s = Solution()
```

```python
print(s.thirdMax(nums = [1 , 33 , 5 , 5 , 1 , 4 , 9]))
```

```python
print(s.thirdMax(nums = [3,2,1]))
```

***Another Optimized Approach***

```python
from typing import List

class Solution:
  def thirdMax(self , nums : List[int]) -> int:
    first = second = third = None

    for num in nums:
      # ignoring duplicates
      if num == first or num == second or num == third:
        continue

      if first is None or num > first:
        third = second
        second = first
        first = num

      elif second is None or num > second:
        third = second
        second = num
      elif third is None or num > third:
        third = num

    return third if third is not None else first

s = Solution()
```

```python
print(s.thirdMax(nums = [12,54,23,1,4,19,20]))
```
