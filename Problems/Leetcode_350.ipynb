```python
from typing import List

class Solution:
  def intersect(self , nums1 : List[int] , nums2 : List[int]) -> List[int]:
    freq = {}
    n1 = len(nums1)
    n2 = len(nums2)

    for i in range(n1):
      freq[nums1[i]] = freq.get(nums1[i] , 0) + 1

    result = []

    for i in range(n2):
      if freq.get(nums2[i] , 0) > 0:
        result.append(nums2[i])
        freq[nums2[i]] = freq[nums2[i]] - 1

    return result

s = Solution()
```

```python
print(s.intersect(nums1 = [4 , 9 , 5] , nums2 = [9,4,9,8,4]))
```
