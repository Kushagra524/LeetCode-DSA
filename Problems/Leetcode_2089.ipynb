```python
from typing import List

class Solution:
  def targetIndices(self , nums : List[int] , target : int) -> List[int]:
    # phase 1: Sorting using Counting Sort ( T.C = O(n + k) )
    n = len(nums)
    maxElement = max(nums)

    freq = [0] * (maxElement + 1)

    for i in nums:
      freq[i] = freq[i] + 1

    nums = []
    for i in range(0 , maxElement + 1):
      while freq[i] > 0:
        nums.append(i)
        freq[i] = freq[i] - 1

    # phase 2: Traversing into sorted Array
    indices = []
    for i in range(n):
      if nums[i] == target:
        indices.append(i)

    return indices

s = Solution()
print(s.targetIndices(nums = [1,2,5,2,3], target = 3))
```
