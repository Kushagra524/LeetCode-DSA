```python
from typing import List

class Solution:
  def savePeople(self , people : List[int] , limit : int) -> int:
    people.sort()
    left = 0
    right = len(people) - 1
    boats = 0

    while left <= right:
      if people[left] + people[right] <= limit:
        left = left + 1

      right = right - 1
      boats = boats + 1

    return boats

s = Solution()
```

```python
print(s.savePeople(people = [1 , 2] , limit = 3))
```

```python
print(s.savePeople(people = [3 , 2 , 2 , 1] , limit = 3))
```

```python
print(s.savePeople(people = [3 , 5 , 3 , 4] , limit = 5))
```
