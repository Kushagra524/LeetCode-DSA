```python
class Solution:
  def reverseVowel(self , s : str) -> str:
    l = list(s)
    left = 0
    right = len(l) - 1
    vowel_range = ['a' , 'e' , 'i' , 'o' , 'u' , 'A' , 'E' , 'I' , 'O' , 'U']

    while left < right:
      while left < right and l[left] not in vowel_range:   # until left is not a vowel and left < right , we increase left
        left = left + 1

      while left < right and l[right] not in vowel_range:   # until right is not a vowel and left < right , we decrease right
        right = right - 1

      if left < right:    # if vowel found which is the only option , we swap them
        l[left] , l[right] = l[right] , l[left]

      left = left + 1   # after swapping we again move left and right pointers
      right = right - 1

    return "".join(l)

s = Solution()
```

```python
print(s.reverseVowel(s = 'IceCrEam'))
```

### Alternative Optimized Approach: Collect, Reverse, and Replace

```python
class SolutionAlternative:
  def reverseVowels(self, s: str) -> str:
    vowels = 'aeiouAEIOU'
    s = list(s)
    left = 0
    right = len(s) - 1
    while left < right:
      while left < right and s[left] not in vowels:
        left = left + 1

      while left < right and s[right] not in vowels:
        right = right - 1

      if left < right:
        s[left] , s[right] = s[right] , s[left]

      left = left + 1
      right = right - 1

    return "".join(s)

s_alt = SolutionAlternative()

print(s_alt.reverseVowels(s = 'IceCream'))
print(s_alt.reverseVowels(s = 'Hello World'))
```
