```python
from typing import List

class Solution:
  def longestSubseq(self , nums : List[int]) -> int:
    xor = 0
    n = len(nums)

    for i in range(n):
      xor = xor ^ nums[i]

    if xor != 0:
      return n

    elif any(nums):
      return n - 1

    return 0

s = Solution()
```

```python
print(s.longestSubseq(nums = [2,3,4,5]))
```

```python
print(s.longestSubseq(nums = [1,2,3,4,5]))
```

```python
print(s.longestSubseq(nums = [1,2,3]))
```
