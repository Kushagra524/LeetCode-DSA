````python
class Solution:
  def addBinary(self , a : str , b : str) -> str:
    na = int(a , 2)
    nb = int(b , 2)

    return f"{(na + nb):b}"   # returnse the binary string of the sum of numbers a and b

s = Solution()
````

````python
print(s.addBinary(a = "1101" , b = "101"))
````

### Explanation of the Optimized `addBinary` Method

This optimized approach simulates manual binary addition, which is often preferred in coding interviews to demonstrate a fundamental understanding of binary arithmetic.

**Class `SolutionOptimized` and Method `addBinary` Initialization:**

```python
class SolutionOptimized:
    def addBinary(self, a: str, b: str) -> str:
        result = []
        i, j = len(a) - 1, len(b) - 1
        carry = 0
```
1.  **`result = []`**: An empty list named `result` is created. This list will store the bits of the sum, building them from the least significant bit (rightmost) to the most significant bit (leftmost).
2.  **`i, j = len(a) - 1, len(b) - 1`**: Two pointers, `i` and `j`, are initialized. They point to the last character (which represents the least significant bit) of binary strings `a` and `b`, respectively. We start processing from the right, just like manual addition.
3.  **`carry = 0`**: A variable `carry` is initialized to `0`. This variable will hold any carry-over value that needs to be added to the next higher bit position.

**The Main Addition Loop:**

```python
        while i >= 0 or j >= 0 or carry:
            sum_val = carry
            if i >= 0:
                sum_val += int(a[i])
                i -= 1
            if j >= 0:
                sum_val += int(b[j])
                j -= 1

            result.append(str(sum_val % 2))
            carry = sum_val // 2
```
This `while` loop continues as long as there are digits remaining in either string `a` or `b`, or if there's a `carry` that still needs to be processed. This ensures all bits and any final carry are handled.

1.  **`sum_val = carry`**: At the beginning of each iteration, `sum_val` is reset to the current `carry`. This accumulates the carry from the previous bit position.
2.  **`if i >= 0: sum_val += int(a[i]); i -= 1`**: If pointer `i` is still within the bounds of string `a`, the integer value of the bit `a[i]` is added to `sum_val`. Then, `i` is decremented to move to the next bit to the left.
3.  **`if j >= 0: sum_val += int(b[j]); j -= 1`**: Similarly, if pointer `j` is within the bounds of string `b`, the integer value of `b[j]` is added to `sum_val`. Then, `j` is decremented.
4.  **`result.append(str(sum_val % 2))`**: After summing `a[i]`, `b[j]`, and the `carry`, `sum_val` can be 0, 1, 2, or 3:
    *   `0 % 2 = 0`
    *   `1 % 2 = 1`
    *   `2 % 2 = 0` (e.g., 1 + 1 = 0 with carry 1)
    *   `3 % 2 = 1` (e.g., 1 + 1 + carry = 1 with carry 1)
    The remainder `sum_val % 2` gives the current bit of the result. This bit is converted to a string and appended to the `result` list.
5.  **`carry = sum_val // 2`**: The integer division `sum_val // 2` calculates the new `carry` for the next higher bit position:
    *   `0 // 2 = 0`
    *   `1 // 2 = 0`
    *   `2 // 2 = 1`
    *   `3 // 2 = 1`

**Final Result Construction:**

```python
        return "".join(result[::-1])
```
1.  **`result[::-1]`**: After the loop finishes, the `result` list contains the binary digits of the sum, but they are in reverse order (least significant bit first). `[::-1]` creates a reversed copy of the list.
2.  **`"".join(...)`**: The `join()` method concatenates all the string elements in the reversed list into a single string, forming the final binary sum.

This method is efficient as it processes the strings bit by bit, directly mimicking how humans perform binary addition.

````python
class SolutionOptimized:
    def addBinary(self, a: str, b: str) -> str:
        result = []
        i, j = len(a) - 1, len(b) - 1
        carry = 0

        while i >= 0 or j >= 0 or carry:
            sum_val = carry
            if i >= 0:
                sum_val += int(a[i])
                i -= 1
            if j >= 0:
                sum_val += int(b[j])
                j -= 1

            result.append(str(sum_val % 2))
            carry = sum_val // 2

        return "".join(result[::-1])


s_optimized = SolutionOptimized()
print(s_optimized.addBinary(a="1101", b="101"))
print(s_optimized.addBinary(a="11", b="1"))
print(s_optimized.addBinary(a="1010", b="1011"))
````
