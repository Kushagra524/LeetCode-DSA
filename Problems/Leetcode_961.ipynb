```python
from typing import List

class Solution:
  def repeatedNTimes(self , nums : List[int]) -> int:
    len_nums = len(nums)

    freq = {}
    for i in nums:
      freq[i] = freq.get(i , 0) + 1

    for i in freq:
      if freq[i] >= (len_nums // 2):
        return i

    return None

s = Solution()
print(s.repeatedNTimes(nums = [5,1,5,2,5,3,5,4]))
```
