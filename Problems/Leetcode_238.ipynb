***Pattern*** : The pattern I saw here is prefix product (similar to prefix sum)

```python
from typing import List

class Solution:
  def productExceptSelf(self , nums : List[int]) -> List[int]:
    n = len(nums)
    answer = [1] * n

    prefix = 1
    for i in range(n):
      answer[i] = prefix
      prefix = prefix * nums[i]

    suffix = 1
    for i in range(n - 1 , -1 , -1):
      answer[i] = answer[i] * suffix
      suffix = suffix * nums[i]

    return answer

s = Solution()
```

```python
print(s.productExceptSelf(nums = [1,2,3,4]))
```

```python
print(s.productExceptSelf(nums = [-1,1,0,-3,3]))
```

***Another Approach***

```python
from typing import List
import math

class Sol:
  def productExceptSelf(self , nums : List[int]) -> List[int]:
    if 0 in nums:
      answer = [0] * len(nums)
      if nums.count(0) > 0:
        zero_id = nums.index(0)
        nums.remove(0)
        answer[zero_id] = math.prod(nums)
      return answer

    return [math.prod(nums) // i for i in nums]

s = Sol()
```

```python
print(s.productExceptSelf(nums = [-1,1,0,-3,3]))
```
