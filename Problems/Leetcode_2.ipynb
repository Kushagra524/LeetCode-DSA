```python
from ast import List
class ListNode:
  def __init__(self ,  val = 0 , next = None):
    self.val = val
    self.next = None

class Solution:
  def linkedListHelper(self , arr1 , arr2):
    def build_list(arr):
      head = ListNode(arr[0])
      curr = head

      for data in arr[1 : ]:
        curr.next = ListNode(data)
        curr = curr.next

      return head
    head1 = build_list(arr1)
    head2 = build_list(arr2)

    return head1 , head2

  def addNums(self , head1 : ListNode | None , head2 : ListNode | None) -> ListNode | None:
    curr1 = head1
    curr2 = head2
    c = 0
    ans = ListNode(-1)
    curr3 = ans

    while curr1 != None or curr2 != None:
      total = c
      c = 0
      if curr1 != None:
        total = total + curr1.val
        curr1 = curr1.next

      if curr2 != None:
        total = total + curr2.val
        curr2 = curr2.next

      if total > 9:
        c = 1
        total = total - 10

      newNode = ListNode(total)
      curr3.next = newNode
      curr3 = curr3.next

    if c > 0:
      newNode = ListNode(c)
      curr3.next = newNode
      curr3 = curr3.next

    return ans.next

  def run(self , arr1 , arr2):
    head1 , head2 = self.linkedListHelper(arr1 , arr2)
    res = self.addNums(head1 , head2)

    return res

s = Solution()
result_node = s.run(arr1 = [1,2,5] , arr2 = [1,1,0])

# Helper to display the result list
output = []
while result_node:
  output.append(result_node.val)
  result_node = result_node.next
print("Result List:", output)
```
