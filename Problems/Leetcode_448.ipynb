***First Approach***

```python
from typing import List

class Solution:
  def findDisappearedNumbers(self , nums : List[int]) -> List[int]:
    freq = [0] * (len(nums) + 1)

    for x in nums:
      freq[x] = freq[x] + 1

    return[x for x in range(1 , len(nums) + 1) if freq[x] == 0]

s = Solution()
```

```python
print(s.findDisappearedNumbers(nums = [4,3,2,7,8,2,3,1]))
```
