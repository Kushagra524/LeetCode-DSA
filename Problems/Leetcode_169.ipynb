```python
from typing import List
import collections

class Solution:
  def majority(self , nums : List[int]) -> int:
    n = len(nums)
    freq = collections.Counter(nums)

    for num in freq:
      if freq[num] > n // 2:
        return num

s = Solution()
```

```python
print(s.majority(nums = [3 , 2 , 3]))
```

```python
print(s.majority(nums = [2 ,2 , 1 , 1 , 2 , 2 , 3]))
```
