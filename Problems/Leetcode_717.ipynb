```python
from typing import List

class Solution:
  def isOneBitCharacter(self , bits : List[int]) -> bool:
    i = 0
    n = len(bits)

    while i < n - 1:
      if bits[i] == 0:   # means we got a 1 bit character
        i = i + 1
      else:   # if bits[i] == 1 : means we got a 2 bit character
        i = i + 2

    return i == (n - 1)

s = Solution()
```

```python
print(s.isOneBitCharacter(bits = [1,0,1,0,1,1,0]))
```

```python
# Dry run with 3 different types of binary arrays
test_cases = {
    "Case 1: Ends with 1-bit character (0)": [1, 1, 1, 0, 0],
    "Case 2: Ends with 2-bit character (10)": [1, 1, 1, 0],
    "Case 3: All single 1-bit characters": [0, 0, 0, 0]
}

for description, bits in test_cases.items():
    result = s.isOneBitCharacter(bits)
    print(f"{description}\nInput: {bits}\nResult: {result}\n{'-'*40}")
```

# 1-Bit and 2-Bit Characters Decoder

This notebook contains a solution to determine if the last character in a binary sequence must be a **one-bit character**.

### The Rules
* **1-bit character:** Must be `0`
* **2-bit character:** Must start with `1` (either `10` or `11`)

---

### How the Algorithm Works (`Solution.isOneBitCharacter`)

We use a pointer $i$ starting at index `0` to traverse the list up to the second-to-last element ($n - 2$):

1. **If `bits[i] == 0`**:
   * This bit must represent a single `0` character.
   * We advance the pointer by **1 step** (`i += 1`).
2. **If `bits[i] == 1`**:
   * This bit must start a double-bit character (`10` or `11`).
   * We advance the pointer by **2 steps** (`i += 2`) to skip the next bit since it is part of this pair.

### The Final Verification
Once the loop completes, we check if our pointer `i` has landed exactly on the very last element of the list (`n - 1`).
* **If `i == n - 1`**: The final `0` stands alone as a 1-bit character $\rightarrow$ returns `True`.
* **If `i == n`**: The final `0` was consumed as the second half of a 2-bit character $\rightarrow$ returns `False`.

---

### Trace Example: `bits = [1, 0, 1, 0, 1, 1, 0]`

* **Initial State**: $i = 0$, $n = 7$
* **Step 1**: `bits[0]` is `1` $\rightarrow$ jump 2 steps $\rightarrow$ $i = 2$
* **Step 2**: `bits[2]` is `1` $\rightarrow$ jump 2 steps $\rightarrow$ $i = 4$
* **Step 3**: `bits[4]` is `1` $\rightarrow$ jump 2 steps $\rightarrow$ $i = 6$
* **Loop Ends**: The loop condition is `while i < 6`. Since $i = 6$, the loop terminates.
* **Check**: Does $i$ equal $n - 1$ ($7 - 1 = 6$)? Yes!
* **Result**: `True` (The final `0` is a standalone 1-bit character).
