```python
from typing import List

class Solution:
  def findLHS(self , nums : List[int]) -> int:
    freq = {}   # empty dictionary to store the occurence of numbers

    for i in nums:
      freq[i] = freq.get(i , 0) + 1   # .get() returns returns the value of key if present otherwise returns the default value (i.e = 0)  , and then + 1 for storing the value as 1 , initially
    max_length = 0

    for ch in freq:
      if ch + 1 in freq:
        current_len = freq[ch] + freq[ch + 1]   # adds the occurences of both the numbers
        max_length = max(max_length , current_len)   # keeps whichever is bigger

    return max_length

s = Solution()
```

```python
print(s.findLHS(nums = [1,3,2,2,5,2,3,7]))
```
