```python
class SOlution:
  def reverseBits(self , n : int) -> int:
    result = 0

    for digit in range(32):
      bit = n & 1      # this gives the last bit of n

      result = (result << 1 ) | bit    # we are shifting the result to 1 place left , and adding 1 bit to the result using | (bitwise OR)

      n = n >> 1   # shifting n (orignal number to right)

    return result

s = SOlution()
```
