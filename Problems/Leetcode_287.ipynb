***First approach*** - *frequency map*

```python
import collections
from typing import List

class Solution:
  def findDuplicate(self , nums : List[int]) -> int:
    if len(nums) == 1:
      return None

    freq = collections.Counter(nums)

    for num in freq:
      if freq[num] >= 2:
        return num

s = Solution()
```

```python
print(s.findDuplicate(nums = [1 , 3 , 4 , 2 , 2]))
```

```python
print(s.findDuplicate(nums = [3 , 1 , 3 , 4 , 2]))
```

```python
print(s.findDuplicate(nums = [3 , 3 , 3 , 3 , 3]))
```

***Second Approach*** - *using hash set , if element not present in set ==> add it to set , is it is already present in set , that's the duplicate element.*

```python
class OptmizedSol:
  def findDuplicate(self , nums : List[int]) -> int:
    set1 = set()

    for num in nums:
      if num in set1:
        return num

      else:
        set1.add(num)

opt = OptmizedSol()
```

```python
print(opt.findDuplicate(nums = [3 , 1 , 3 , 4 , 2]))
```

***Third approach*** - *using regular foor loop and .count() method , but it may return TLE in case of large list*

```python
from typing import List

class Solutions:
  def findDuplicate(self , nums : List[int]) -> int:
    # This approach is O(n^2) because .count() scans the list inside a loop.
    # How could we use a 'set' to keep track of seen numbers in O(n) time?
    for i in range(len(nums)):
      if nums.count(nums[i]) >= 2:
        return nums[i]


sol = Solutions()
```

```python
print(sol.findDuplicate(nums = [3 , 1 , 3 , 4 , 2]))
```
