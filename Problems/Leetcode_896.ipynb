```python
from typing import List

class Solution:
  def isMonotonic(self , nums : List[int]) -> bool:
    increasing = True
    decreasing = True

    for i in range(len(nums) - 1):
      if nums[i] > nums[i + 1]:
        decreasing = False

      if nums[i] < nums[i + 1]:
        increasing = False

    return increasing or decreasing

s = Solution()
```

```python
print(s.isMonotonic(nums = [1,3,2]))
```

```python
print(s.isMonotonic(nums = [1,2,2,3]))
```
