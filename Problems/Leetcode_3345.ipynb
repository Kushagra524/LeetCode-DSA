```python
class Solution:
  def smallestDiv(self , n : int , t : int) -> int:
    while True:
      product = 1
      curr_num = n

      while curr_num > 0:
        product = product * (curr_num % 10)
        curr_num = curr_num // 10

        if product % t == 0:
          return n

        n = n + 1

s = Solution()
```

```python
print(s.smallestDiv(n = 10 , t = 2))
```
