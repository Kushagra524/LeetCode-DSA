**1st Approach - Normal Logic**

```python
from typing import List

class Solution:
  def findLucky(self , arr : List[int]) -> int:
    maxLucky = 0
    for i in arr:
      if i == arr.count(i):
        maxLucky = max(i , maxLucky)

    return maxLucky if maxLucky > 0 else -1

s = Solution()
```

```python
print(s.findLucky(arr = [2,2,3,4]))
```

**2nd Approach - better time complexity**

```python
from typing import List

class Sol:
  def findLucky2(self , arr : List[int]) -> int:
    hash_map = {}
    largest = -1

    for i in arr:
      hash_map[i] = hash_map.get(i , 0) + 1

    for i in hash_map:
      if hash_map[i] == i and i > largest:
        largest = i

    return largest

s = Sol()
```

```python
print(s.findLucky2(arr = [2,2,3,4]))
```
