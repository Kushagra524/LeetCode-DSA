***First Approach*** - using HashMap takes O(i * j) time complexity in case of i-rows and j-columns

```python
from typing import List

class Solution:
  def intersection(self , nums : List[List[int]]) -> List[int]:
    n = len(nums)
    freq = {}
    result = []

    for row in nums:
      for x in row:
        freq[x] = freq.get(x , 0) + 1

    for row in nums:
      for x in row:
        if freq[x] == len(nums):
          result.append(x)
          freq[x] = freq[x] - 1

    result.sort()

    return result

s = Solution()
```

```python
print(s.intersection(nums = [[3,1,2,4,5],[1,2,3,4],[3,4,5,6]]))
```

***Second Approach*** - using set  works in O(n)

```python
from typing import List

class Sol:
  def intersection(self , nums : List[List[int]]) -> List[int]:
    set1 = set(nums[0])

    for row in nums:
      set1 = set1.intersection(set(row))

    l1 = list(set1)
    l1.sort()

    return l1

sol = Sol()
```

```python
print(sol.intersection(nums = [[3,1,2,4,5],[1,2,3,4],[3,4,5,6]]))
```
