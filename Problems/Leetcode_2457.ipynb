# Problem Explanation

You are given a positive integer n and a target value target. A number is called beautiful if the sum of its digits is less than or equal to target. Your task is to find the smallest non-negative integer x that can be added to n such that the new number (n + x) becomes beautiful. If n is already beautiful, then no addition is needed, so the answer is 0. The challenge is to find the minimum possible addition, not just any addition, that makes the digit sum satisfy the given condition

# **Algorithm**

```python
class Solution:
  def digitsSum(self , n : int):
    total = 0

    while n > 0:
      total = total + (n % 10)
      n = n // 10

    return total

  def makeIntegerBeautiful(self , n : int , target : int) -> int:
    orignal = n
    place = 10

    while self.digitsSum(n) > target:
      add = place - (n % place)
      n = n + add
      place = place * 10


    return n - orignal

s = Solution()
```

```python
print(s.makeIntegerBeautiful(n = 16 , target = 6))   # output = 4 , 16 + 4 = 2 ===> 2 + 0 => 2 < target i.e. 2 < 6
```
