***First approach*** - Dynamic Sliding with Hash Map

```python
class Solution:
  def longestSub(self , s : str) -> int:
    left = 0
    max_len = 0
    freq = {}

    for right in range(len(s)):
      freq[s[right]] = freq.get(s[right] , 0) + 1

      while freq[s[right]] > 1:
        freq[s[left]] = freq[s[left]] - 1
        left = left + 1

      window_len = right - left + 1
      max_len = max(max_len , window_len)

    return max_len

s = Solution()
```

```python
print(s.longestSub(s = "ababbahab"))
```

***Second Approach*** - Dynamic Sliding Window with Hash Set

```python
class Sol:
  def longestSubSeq(self , s : str) -> int:
    left = 0
    seen = set()
    max_len = 0

    if len(s) == 1:
      return 1

    for right in range(len(s)):
      while s[right] in seen:
        seen.remove(s[left])
        left = left + 1

      seen.add(s[right])
      max_len = max(max_len , right - left + 1)

    return max_len

sol = Sol()
```

```python
print(sol.longestSubSeq(s = "abbahaja"))
```
