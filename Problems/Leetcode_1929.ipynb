***First Approach***

```python
from typing import List

class Solution:
  def concatenateArray(self , nums : List[int]) -> List[int]:
    for i in range(len(nums)):
      nums.append(nums[i])

    return nums

s = Solution()
```

```python
print(s.concatenateArray(nums = [1 , 2 , 3 , 1]))
```

***Second Approach***

```python
class Solution:
  def concatArray(self , nums : List[int]) -> List[int]:
    return nums + nums

s1 = Solution()
```

```python
print(s1.concatArray(nums = [1 , 2 , 1 , 3]))
```
