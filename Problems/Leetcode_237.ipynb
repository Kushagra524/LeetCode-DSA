```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val = 0 , next = None):
        self.val = val
        self.next = next

class Solution:
    def linkedListHelper(self , arr):
      # to convert my Python List into a linked list
      if not arr:
        return None
      head = ListNode(arr[0])
      curr = head

      for val in arr[1 : ]:
        curr.next = ListNode(val)
        curr = curr.next

      return head      # return

    def deleteNode(self, node):
        """
        :type node: ListNode
        :rtype: void Do not return anything, modify node in-place instead.
        """
        # 'node' is the node I wanna delete
        node.val = node.next.val     # node's value set same as next node's value
        node.next = node.next.next   # then we connect this node to the next next node , so , that particular node gets deleted

    def printResult(self , head):
      res = []
      curr = head

      while curr:
        res.append(curr.val)
        curr = curr.next

      print(res)


    def run(self , arr , target):
      head = self.linkedListHelper(arr)
      node = head

      while node and node.val != target:
        node = node.next


      if node is None or node.next is None:
        return

      deleted_value = node.val
      self.deleteNode(node)
      print(deleted_value)


s = Solution()
arr = [1,2,3,4,5,6]
target = 5
s.run(arr , target)
```
