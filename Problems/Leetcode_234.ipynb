```python
class ListNode:
  def __init__(self , val = 0 , next = None):
    self.val = val
    self.next = next


class Solution:
  def linkedListHelper(self , arr):
    head = ListNode(arr[0])
    curr = head

    for data in arr[1 : ]:
      curr.next = ListNode(data)
      curr = curr.next

    return head

  def isPalindrome(self , head : ListNode | None) -> bool:
    if head is None or head.next is None:
      return True

    # first I will reach the middle node using two pointers
    slow = head
    fast = head

    while fast != None and fast.next != None:
      slow = slow.next
      fast = fast.next.next

    # Then the core logic is to reverse 2nd half of the Linked List and then compare it with 1st half => if same return True else: return False
    curr = slow
    prev = None
    nxt = None

    while curr != None:
      nxt = curr.next
      curr.next = prev
      prev = curr
      curr = nxt

    p1 = prev
    p2 = head

    while p2 != None:
       # most of us will think that we could check for p1 != None , but but but , in case if oginal linked list is of odd length , then p1 would contains 1 node more than p2 and then while checking p1 != None , cwe would reach end of p1 but in that case p2's last node will be None and companring p1.val with p2.val (i.e. p2.val = None) , means it will return false unnecesarrily
      if p1.val != p2.val:
        return False

      p1 = p1.next
      p2 = p2.next

    return True

  def run(self , arr):
    head = self.linkedListHelper(arr)
    res = self.isPalindrome(head)

    return res

s = Solution()
print(s.run(arr = [1,2,3,2,1]))
```
