```python
from typing import List

class Solution:
  def smallestIndex(self , nums : List[int]) -> int:
    n = len(nums)
    for i in range(n):
      sum = 0
      while nums[i] > 0:
        sum = sum + (nums[i] % 10)
        nums[i] = nums[i] // 10

      if sum == i:
        return i

    return -1

s = Solution()
print(s.smallestIndex(nums = [1,10,11]))   # 1 because at index = 1 , the sum of digits of 10 = 1 + 0 ==> 1 , so the condition of ==> (sum(nums[i]) == i) fulfilled successfully.
```
