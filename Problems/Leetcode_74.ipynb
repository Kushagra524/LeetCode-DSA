```python
from typing import List

class Solution:
  def searchMatrix(self , matrix : List[List[int]] , target : int) -> bool:
    rows = len(matrix)
    cols = len(matrix[0])
    l = 0
    r = (rows * cols) - 1

    while l <= r:
      mid = (l + r) // 2

      row_pos = mid // cols
      col_pos = mid % cols

      if matrix[row_pos][col_pos] == target:
        return True

      elif matrix[row_pos][col_pos] > target:
        r = mid - 1

      else:
        l = mid + 1

    return False


s = Solution()
print(s.searchMatrix(matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]], target = 3))
```
