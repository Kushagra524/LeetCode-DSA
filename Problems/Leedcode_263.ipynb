```python
class Solution:
  def detectUglyNum(self , num : int) -> bool:
    # first I'll remove all 2's until and unless their are no 2's left , then remove all 3's and then remove all 5's , if at last we get a number that isn't == 1 , that number is not ugly (False) , otherwise Ugly (True)

    while num % 2 == 0:   # condition of removing 2's
      num = num // 2

    while num % 3 == 0:  # condition for removing all 3's
      num = num // 3

    while num % 5 == 0:
      num = num // 5

    if num == 1:
      return True
    else:
      return False

s = Solution()
```

```python
print(s.detectUglyNum(num = 14))   # 2 * 7 = 14
```

```python
print(s.detectUglyNum(num = 24))
```
