```python
from typing import List

class Solution:
  def maximumGap(self , nums : List[int]) -> int:
    n = len(nums)
    if n < 2:
      return 0
    nums.sort()

    max_gap = 0
    for i in range(n - 1):
      curr_gap = nums[i + 1] - nums[i]
      if curr_gap > max_gap:
        max_gap = curr_gap

    return max_gap

s = Solution()
print(s.maximumGap(nums = [12 , 10 , 45 , 51]))
```
