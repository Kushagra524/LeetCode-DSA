```python
from typing import List

class Solution:
  def validParenthesis(self , s : str) -> bool:
    stack = []

    for br in s:
      if br == '(' or br == '{' or br == '[':
        stack.append(br)

      else:
        if not stack:
          return False

        if br == ')' and stack[-1] != '(':
          return False

        elif br == '}' and stack[-1] != '{':
          return False

        elif br == ']' and stack[-1] != '[':
          return False

        else:
          stack.pop()

    return len(stack) == 0

s = Solution()
print(s.validParenthesis(s = '(()[{[(]}])'))
```
