```python
class Solution:
  def strStr(self , needle : str , haystack : str) -> int:
    n = len(haystack)
    m = len(needle)

    if m == 0:
      return 0

    for i in range(n - m + 1):
      if haystack[i : i + m] == needle:
        return i

    return -1

s = Solution()
```

```python
print(s.strStr(needle = "but" , haystack = "sadbutsad"))
```
