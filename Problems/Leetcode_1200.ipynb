```python
from typing import List

class Solution:
  def minimumAbsDifference(self , arr : List[int]) -> List[int]:
    n = len(arr)
    arr.sort()
    res = []
    min_dif = float('inf')

    for i in range(0 , n - 1):
      diff = arr[i + 1] - arr[i]

      if diff < min_dif:
        min_dif = diff
        res = [[arr[i] , arr[i + 1]]]

      elif diff == min_dif:
        res.append([arr[i] , arr[i + 1]])

    return res

s = Solution()
print(s.minimumAbsDifference(arr = [4 , 2 , 1 , 3]))
```
