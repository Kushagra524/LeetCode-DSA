```python
class Solution:
  def maxDepthParenthesis(self , s : str) -> int:
    max_depth = 0
    curr_depth = 0
    # I just simply need to calculate the maximum times , '(' opens up , so that will be our maximum depth.
    for ch in s:
      if ch == '(':
        curr_depth = curr_depth + 1

      if max_depth < curr_depth:
        max_depth = curr_depth

      if ch == ')':
        curr_depth = curr_depth - 1

    max_depth = max(max_depth , curr_depth)

    return max_depth

s = Solution()
print(s.maxDepthParenthesis("(1+(2*3)+((8)/4))+1"))
```
