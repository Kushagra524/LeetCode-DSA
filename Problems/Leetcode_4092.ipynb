```python
class Solution:
  def binaryDigitScore(self , a : int , b : int) -> int:
    s1 = 0
    s2 = 0

    while a:
      s1 = s1 + (a & 1)
      a = a >> 1

    while b:
      s2 = s2 + (b & 1)
      b = b >> 1

    if s1 >= s2:
      return s1 * s2

    else:
      return s2 // s1

s = Solution()
print(s.binaryDigitScore(a = 5 , b = 15))
```
