# *First approach*

```python
from typing import List

class Solution:
  def removeElement(self , nums : List[int] , val : int) -> int:

    i = 0
    while i < len(nums):
      if nums[i] == val:
        nums.pop(i)

      else:
        i = i + 1
    return len(nums)

s = Solution()
```

```python
print(s.removeElement(nums = [3 , 2 , 2 , 3] , val = 3))
```

# *Second Approach*

```python
from typing import List

class Solution:
  def removeElement(self , nums : List[int] , val : int) -> List[int]:
    k = 0

    for i in range(len(nums)):
      if nums[i] != val:           # if the i-th element is not equal to val , then ..
        nums[k] = nums[i]          # nums[k] gets the value of nums[i]
        k = k + 1  # denotes the position(index) where the current "i" is , but actually returns the number of elements in the array which are not equal to val

    return k

s = Solution()

# nums = [3 , 2 , 2 , 3] , val = 3

# k = 0 and i = 0  --> checking nums[0] != val  => 3 equal to 3 , skip

# k = 0 and i = 1 --> checking nums[1] != val => 2 not equal to 3 so, putting nums[k] = nums[i] ==> nums[0] = nums[1] ==> at 0-th index we put 2 , so array becomes [2 , 2 , 2 , 3] and k = k + 1

# k = 1 and i = 2 , 2 not eqaul 3 , nums[k i.e 1]  = nums[i i.e 2] , 1-st index gets values = 2 , array becomes [2 , 2 , 2 , 3] and k = k + 1 --> k = 2

# k = 2 and i = 3 , nums[3] equal to 3 , so skip

# out k is equal to 2 , means we 2 elements were not eqaul to val
```

```python
print(s.removeElement(nums = [3 , 2 , 2 , 3] , val = 3))
```
