***Approach***

1. I can do this with the help of hash_map , the first we all know is that if we sort anagrams strings , we get two same strings.

2. I will make an empty hash_map = {}

3. Also I will sort each string in my list and its sorted version of each string will called as key.

4. Then I traverse through each element(string) of my list and check if the particular key(sorted version of that key) is present in my hash_map ??. If yes , it's sorted part becomes the value of that partiula

```python
from typing import List

class Solution:
  def sortString(self , s):
    s1 = list(s)   # for example s = "eat" ==> s1 becomes ["e" , "a" , "t"]

    s1.sort()    # s1 = ["a" , "e" , "t"]

    return "".join(s1)  # s1 = "aet" which is returned into the function below becoz it is called inside the below funtion and it is the key

  def groupAnagram(self , strs : List[str]) -> List[List[str]]:
    hash_map = {}

    for s in strs:
      key = self.sortString(s)  # first I will sort the string , and make it a key  ,so my key is "aet"

      if key in hash_map:    # if "eat" present in hash_map , it's value = "eat" whch is my orignal string (s) will be appended into "aet"
        hash_map[key].append(s)

      else:
        hash_map[key] = [s] # if that key is not present than we initialize with a list containing the string
    # print(hash_map)
    return list(hash_map.values())

s = Solution()
```

```python
print(s.groupAnagram(strs = ['eat' , 'tea' , 'tan' , 'ate' , 'nat' , 'bat']))
```
