```python
class Solution:
  def perfectNumber(self , num : int) -> bool:
    check = 1
    if num == 1:
      return False

    for i in range(2 , int(num ** 0.5) + 1):
      if num % i == 0:
        check = check + i

        if i != num // i:
          check = check + num // i

    return check == num

s = Solution()
```

```python
print(s.perfectNumber(num = 28))
```
