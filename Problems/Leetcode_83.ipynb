**Main logic is simply changing the links becoz linked list is combination of elements linked together**

```python
class ListNode:
  def __init__(self, val=0, next=None):
    self.val = val
    self.next = next

class Solution:
  def linkedListHelper(self, arr):
    if not arr:
      return None

    head = ListNode(arr[0])
    curr = head

    for val in arr[1 : ]:
      curr.next = ListNode(val)
      curr = curr.next

    return head

  def removeDuplicates(self, head: ListNode | None) -> ListNode | None:
    p1 = head
    if p1 == None or p1.next == None:
      return p1

    while p1.next != None:
      if p1.val == p1.next.val:
        p1.next = p1.next.next
      else:
        p1 = p1.next

    return head

  def run(self, arr):
    head = self.linkedListHelper(arr)
    newhead = self.removeDuplicates(head)
    curr = newhead

    while curr:
      print(curr.val , end = " ")
      curr = curr.next
    return newhead

s = Solution()
res = s.run(arr=[2,2,5,5,5,6,7])
```
