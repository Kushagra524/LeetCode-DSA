```python
import collections

class Solution:
  def findDiff(self , s : str , t : str) -> str:
    freq = collections.Counter(t)

    for i in s:
      freq[i] = freq[i] - 1

    for i in freq:
      if freq[i] == 1:
        return i

s = Solution()
```

```python
print(s.findDiff(s = "a" , t = "aa"))
```
