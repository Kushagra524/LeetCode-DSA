***First approach***

```python
# nums = [1,2,3,4,5,6,7], k = 3
# output = [5 , 6 , 7 , 1 , 2 , 3 , 4]
from typing import List

class Sol:
  def rotateArray(self , nums : List[int] , k : int) -> List[int]:
    n = len(nums)
    k = k % n

    nums.reverse()
    nums[k : ] = reversed(nums[k : ])
    nums[ : k] = reversed(nums[ : k])

    return nums


s = Sol()
print(s.rotateArray(nums = [1,2,3,4,5,6,7], k = 3))
print(s.rotateArray(nums = [-1,-100,3,99], k = 2))
```

***Second approach (Optimized)***

```python
from typing import List

class Sol:
  def rotateArray(self , nums : List[int] , k : int) -> List[int]:
    n = len(nums)
    k = k % n  # rotating k times is same as rotating n - k time , in case k is very large

    if k > 0:
      right_part = nums[ n - k : ]   # seperating right part
      del nums[n - k : ] # deleting the right part from orignal array
      nums[0 : 0] = right_part  # nums[0 : 0] creates an empty slice where I'm placing the right part

    return nums

s = Sol()
print(s.rotateArray(nums = [1,2,3,4,5,6,7], k = 3))
print(s.rotateArray(nums = [-1,-100,3,99], k = 2))
```
