# **Logic**

```python
# I need to divide the array into 3 non empty continous parts such that sum of all parts is equal.

# arr = [0, 2, 1, -6, 6, -7, 9, 1, 2, 0, 1]

# so we can divide into :
# part1: 0 + 2 +1 = 3, part2: -6 + 6 + -7 + 9 + 1 = 3, part3: 2 + 0 + 1 = 3
```

```python
from typing import List

class Solution:
  def canThreePartsEqualSum(self , arr : List[int]) -> bool:
    total_sum = sum(arr)

    if total_sum % 3 != 0:
      return False

    target = total_sum // 3
    count = 0
    currSum = 0

    for i in arr:
      currSum = currSum + i

      if currSum == target:
        count = count + 1
        currSum = 0

      if count == 3:
        break

    return count == 3

s = Solution()
```

```python
print(s.canThreePartsEqualSum(arr = [0, 2, 1, -6, 6, -7, 9, 1, 2, 0, 1]))
```

```python
print(s.canThreePartsEqualSum(arr = [0,2,1,-6,6,7,9,-1,2,0,1]))
```
