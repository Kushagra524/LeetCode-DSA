```python
from typing import List

class Solution:
  def finalValueAfterOperations(self , operations : List[str]) -> int:
    x = 0

    for i in operations:
      if i == '--X' or i == 'X--':
        x = x - 1
      else:
        x = x + 1

    return x

s = Solution()
```

```python
print(s.finalValueAfterOperations(operations = ["--X","X++","X++"]))
```
