***1. Simple Iteration in Linket List***

image.png

```python
from typing import Optional

class ListNode:
  def __init__(self , val , next = None):
    self.val = val
    self.next = next


class Solution:
  def middleNode(self , head : ListNode | None) -> ListNode | None:
    if not head:
      return None

    curr = head
    l = 0

    while curr != None:
      l = l + 1
      curr = curr.next


    curr = head
    for i in range(l // 2):
      curr = curr.next

    return curr

def linkedListHelper(arr):
  if not arr:
    return None

  head = ListNode(arr[0])
  temp = head

  for val in arr[1 : ]:
    temp.next = ListNode(val)
    temp = temp.next

  return head
```

```python
arr = [1,2,3,4,5,6]
headNode = linkedListHelper(arr)

s = Solution()
middle = s.middleNode(headNode)

# converting my Linked List into Python List again to preview answer , but it's not required on Leetcode to do this stuff.

result = []
curr = middle
while curr:
  result.append(curr.val)
  curr = curr.next

print(result)
```

***2nd approach - Fast & Slow Pointer***

image.png

```python
class ListNode:
  def __init__(self , val = 0 , next = None):
    self.val = val
    self.next = next

class Solution:
  def middleNode(self , head : ListNode | None) -> ListNode | None:
    slow = head
    fast = head

    while fast != None and fast.next != None:
      slow = slow.next
      fast = fast.next.next

    return slow

s = Solution()
print(s.middleNode())
```
