```python
from typing import List

class Solution:
  def containsNearbyDuplicate(self , nums : List[int] , k : int) -> bool:
    n = len(nums)
    window = set()

    for i in range(0 , n):
      if nums[i] in window:
        return True

      window.add(nums[i])

      if len(window) > k:
        window.discard(nums[i - k]) # Changed from remove to discard for robustness , becoz .remove() raises key error if element not found in set , but .discard() doesn't

    return False

s = Solution()
```

```python
print(s.containsNearbyDuplicate(nums = [1,2,3,1,2,3], k = 2))
```
