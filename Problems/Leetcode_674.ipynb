```python
from typing import List

class Solution:
  def findLengthOfLCIS(self , nums : List[int]) -> int:
    count = 1
    max_len = 1
    n = len(nums)

    for i in range(1 , n):
      if nums[i - 1] < nums[i]:
        count = count + 1
      else:
        count = 1

      max_len = max(max_len , count)

    return max_len

s = Solution()
```

```python
print(s.findLengthOfLCIS(nums = [1 , 3 , 5 , 4 , 7]))
```
