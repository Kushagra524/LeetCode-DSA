```python
class ListNode():
  def __init__(self , val = 0 , next = None):
    self.val = val
    self.next = next

class Solution:
  def linkedListHelper(self , arr):
    if not arr:
      return None

    head = ListNode(arr[0])
    curr = head

    for val in arr[1 : ]:
      curr.next = ListNode(val)
      curr = curr.next

    return head

  def reverseLinkedList(self , head : ListNode | None) -> ListNode | None:
    curr = head
    prev = None   # note that prev and nxt are just pointers made by us.
    nxt = None    # nxt is also a pointer but it's not equal to next

    while curr != None:
      nxt = curr.next
      curr.next = prev
      prev = curr
      curr = nxt

    return prev

  def run(self , arr):
    head = self.linkedListHelper(arr)
    newhead = self.reverseLinkedList(head)

    curr = newhead

    while curr != None:
      print(curr.val , end = " ")
      curr = curr.next

    return newhead

s = Solution()
res = s.run(arr = [1,2,3,4,5])
```
