```python
from typing import List

class Solution:
  def findWords(self , words : List[str]) -> List[str]:
    row1 = set("qwertyuiop")
    row2 = set("asdfghjkl")
    row3 = set("zxcvbnm")

    res = []

    for word in words:
      w = word.lower()

      if w[0] in row1:
        row = row1

      elif w[0] in row2:
        row = row2

      else:
        row = row3

      contains = True

      for ch in w:
        if ch not in row:
          contains = False
          break

      if contains:
          res.append(word)
    return res

s = Solution()
```

```python
print(s.findWords(words = ["Hello","Alaska","Dad","Peace"]))
```
