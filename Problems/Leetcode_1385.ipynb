```python
from typing import List

class Solution:
  def findTheDistanceValue(self , arr1 : List[int] , arr2 : List[int] , d : int) -> int:
    count = 0
    length = len(arr1)

    for i in arr1:
      for j in arr2:
        if abs(i - j) <= d:
          count = count + 1
          break

    return length - count

s = Solution()
```

```python
print(s.findTheDistanceValue(arr1 = [4 , 5 , 8] , arr2 = [10 , 9 , 1 , 8] , d = 2))
```
