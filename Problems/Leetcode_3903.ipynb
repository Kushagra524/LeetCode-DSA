```python
from typing import List

class Solution:
  def firstStableIndex(self , nums : List[int] , k : int) -> int:
    for i in range(0 , len(nums)):
      if max(nums[0 : i + 1]) - min(nums[i : ]) <= k:
        return i

    return -1

s = Solution()
print(s.firstStableIndex(nums = [5,0,1,4], k = 3))
```
