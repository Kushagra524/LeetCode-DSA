```python
from typing import List

class Solution:
  def numIdenticalPairs(self , nums : List[int]) -> int:
    count = 0
    n = len(nums)

    for i in range(n):
      for j in range(i + 1 , n):
        if nums[i] == nums[j]:
          count = count + 1

    return count

s = Solution()
```

```python
print(s.numIdenticalPairs(nums = [1,1,2,3,4,1,2,3]))
```
