```python
from typing import List

class Solution:
  def maxConsecutiveOnes(self , nums : List[int]) -> int:
    curr_count = 0
    max_count = 0
    for num in nums:
      if num == 1:
        curr_count = curr_count + 1
        max_count = max(max_count , curr_count)

      else:
        curr_count = 0

    return max_count

s = Solution()
```

```python
print(s.maxConsecutiveOnes(nums = [1,1,0,1,1,1]))
```
