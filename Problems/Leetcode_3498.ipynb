```python
class Solution:
  def reverseDegree(self , s : str) -> int:
    res = 0
    for i in range(len(s)):
      res = res + ((123 - ord(s[i])) * (i + 1))   # main formulae , and it's logic I made using concept of ASCII value.

    return res

s = Solution()
print(s.reverseDegree(s = 'abc'))
```

# Explanation of the 'Reverse Degree' Logic

This algorithm calculates a custom numerical weight (or "degree") for an input string based on two main concepts: **ASCII mapping** and **positional weighting**.

### 1. ASCII Value Calculation
`ord(s[i])` returns the Unicode code point (ASCII value) of the character at index `i`:
* `'a'` $\rightarrow 97$
* `'b'` $\rightarrow 98$
* `'c'` $\rightarrow 99$

### 2. Reverse Alphabetical Scale: `123 - ord(s[i])`
By subtracting the character's ASCII value from `123`, we create an inverted rating where characters closer to the beginning of the alphabet have a higher value:
* For `'a'`: $123 - 97 = 26$
* For `'b'`: $123 - 98 = 25$
* For `'c'`: $123 - 99 = 24$
* For `'z'`: $123 - 122 = 1$

### 3. Positional Weighting: `(i + 1)`
To ensure that the character's position within the string influences the score, the reverse ASCII score is multiplied by its 1-based index `(i + 1)`. Characters positioned later in the string are scaled up significantly more than those at the beginning.

### Example Trace for `s = 'abc'`

| Character | Index ($i$) | Position ($i + 1$) | Reverse ASCII ($123 - \text{ord}(c)$) | Contribution | Sum |
| :--- | :--- | :--- | :--- | :--- | :--- |
| `'a'` | $0$ | $1$ | $123 - 97 = 26$ | $26 \times 1 = 26$ | **26** |
| `'b'` | $1$ | $2$ | $123 - 98 = 25$ | $25 \times 2 = 50$ | $26 + 50 =$ **76** |
| `'c'` | $2$ | $3$ | $123 - 99 = 24$ | $24 \times 3 = 72$ | $76 + 72 =$ **148** |

**Final Result:** `148`
