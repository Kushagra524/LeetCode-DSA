```python
class Solution:
  def addStrings(self , num1 : str , num2 : str) -> str:
    carry = 0
    res = ""

    i = len(num1) - 1
    j = len(num2) - 1

    while i >= 0 or j >= 0 or carry:
      if i >= 0:
        digit1 = int(num1[i])
      else:
        digit1 = 0

      if j >= 0:
        digit2 = int(num2[j])
      else:
        digit2 = 0

      sum = digit1 + digit2 + carry
      res = res + str(sum % 10)
      carry = sum // 10

      i = i - 1
      j = j - 1

    return res[ : : -1]

s = Solution()
```

```python
print(s.addStrings(num1 = "5412" , num2 = "42"))
```
