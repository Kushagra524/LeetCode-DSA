**Best Approach**

```python
from typing import List

class Solution:
  def minMoves(self , nums : List[int]) -> int:
    return max(nums) * len(nums) - sum(nums)

s = Solution()
print(s.minMoves(nums = [2,1,3]))
```

**Another approach**

```python
from typing import List

class Sol:
  def minMoves2(self , nums : List[int]) -> int:
    count = 0
    for i in nums:
      while i < max(nums):
        i = i + 1
        count = count + 1

    return count

s = Sol()
print(s.minMoves2(nums = [2,1,3]))
```
