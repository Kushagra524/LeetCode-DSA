```python
from typing import List

class Solution:
  def countElements(self , nums : List[int]) -> int:
    minimum = min(nums)
    maximum = max(nums)
    count = 0

    for n in nums:
      if n > minimum and n < maximum:
        count = count + 1


    return count

s = Solution()
```

```python
print(s.countElements(nums = [11 , 7 , 2 , 15]))
```
