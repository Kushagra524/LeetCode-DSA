```python
class Solution:
  def reverseOnlyLetters(self, s: str) -> str:
    newS = list(s)
    left = 0
    right = len(newS) - 1

    while left < right:
      if not newS[left].isalpha():
        left += 1
      elif not newS[right].isalpha():
        right -= 1
      else:
        newS[left], newS[right] = newS[right], newS[left]
        left += 1
        right -= 1

    return "".join(newS)

s = Solution()
print(s.reverseOnlyLetters(s = "ab-bkba-h-ccb"))
```
