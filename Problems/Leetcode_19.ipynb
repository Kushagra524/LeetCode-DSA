**First Approach**

```python
class ListNode:
  def __init__(self , val = 0 , next = None):
    self.val = val
    self.next = next

class Solution:
  def linkedListHelper(self , arr):
    if not arr:
      return None

    head = ListNode(arr[0])
    curr = head

    for val in arr[1 : ]:
      curr.next = ListNode(val)
      curr = curr.next

    return head

  def removeNthFromEnd(self , head : ListNode | None , n : int) -> ListNode | None:
    curr = head
    l = 0
    while curr:
      l = l + 1
      curr = curr.next

    # If I need to remove the head node
    if l == n:
      return head.next

    curr = head
    for i in range(l - n - 1):
      curr = curr.next
    curr.next = curr.next.next
    return head

  def run(self , arr , n):
    head = self.linkedListHelper(arr)
    newhead = self.removeNthFromEnd(head , n)

    # Simple print loop inside the run method
    curr = newhead
    while curr:
      print(curr.val , end = " ")
      curr = curr.next

    return newhead

s = Solution()
res = s.run(arr = [1,2,3,4,5,6] , n = 2)
```

```python
from inspect import currentframe
class ListNode:
  def __init__(self , val = 0 , next = None):
    self.val = val
    self.next = next


class Solution:
  def linkedListHelper(self , arr):
    if not arr:
      return None

    head = ListNode(arr[0])
    curr = head

    for val in arr[1 : ]:
      curr.next = ListNode(val)
      curr = curr.next

    return head

  def removeNthFromEnd(self , head : ListNode | None , n : int) -> ListNode | None:
    p1 = head
    p2 = head    # I take two pointers p1 and p2 both starting from head

    # Now imagine what I do is move p2 n-times ahead of while p1 is sleeping.
    for i in range(n):
      p2 = p2.next

    while p2.next != None:
      p1 = p1.next
      p2 = p2.next
      # doing the above while loop results an achievement that now p1 will be exactly at 1 place before where n-th place from end is located and p2 will reach the last node.

    # now , i will connect p1 to it's next next node.
    p1.next = p1.next.next
    return head

  def run(self , arr , n):
    head = self.linkedListHelper(arr)
    newHead = self.removeNthFromEnd(head , n)

    curr = newHead
    while curr:
      print(curr.val , end = " ")
      curr = curr.next

    return newHead

s = Solution()
res = s.run(arr = [1,2,3,4,5,6] , n = 3)
```
