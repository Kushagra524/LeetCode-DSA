```python
# 1. s = "(ed(et(oc))el)"

# 2. "(ed(etco)el)"

# 3. "(edocteel)"

# 4. "leetcode"
```

```python
class Solution:
  def reverseParentheses(self , s : str) -> str:
    stack = []
    curr_string = ""

    for ch in s:
      if ch == '(':
        stack.append(curr_string)   # whenever I see a '(' , I will save the current string into my stack

        curr_string = ""  # and then reset my curr_string to initial

      elif ch == ')':
        curr_string = curr_string[ : : -1]   # wheneever I see a ')' , I will reverse the current string and attach it to the previous string in my stack

        curr_string = stack.pop() + curr_string  # stack.pop() returns the last saved string inside the stack and attaches it with the current string , just like example - stack = ["u" , "love"] , then curr_string = "uevol"

      else:
        curr_string = curr_string + ch   # if characters come , they simply added to the curr_string to proceed further.

    return curr_string

s = Solution()
print(s.reverseParentheses(s = "(u(love)i)"))
```
