```python
from typing import List

class Solution:
  def prefixDivisibleBy5(self , nums : List[int]) -> List[bool]:
    res = []
    curr = 0

    for i in nums:
      curr = curr * 2 + i      # I did curr * 2 + i becoz I need to calculate the divisibility of binary number by 5
      res.append(curr % 5 == 0)

    return res

s = Solution()
```

```python

print(s.prefixDivisibleBy5(nums = [0,1,1]))
```

```python
print(s.prefixDivisibleBy5(nums = [1,1,1]))
```
