```python
from typing import List

class Solution:
  def numberOfEmployeesWhoMetTarget(self , hours : List[int] , target : int) -> int:
    count = 0
    for hr in hours:
      if hr >= target:
        count = count + 1

    return count

s = Solution()
```

```python
print(s.numberOfEmployeesWhoMetTarget(hours = [0,1,2,4,3,1,5] , target = 2))
```
