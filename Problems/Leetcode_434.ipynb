***First Approach*** - using built-in function (simplest and not recommended for interviews or coding assessments)

```python
class Solution:
  def numOfSegments(self , s : str) -> int:
    return len(s.split())

s = Solution()
```

```python
print(s.numOfSegments(s = "Heyy , My name is Ayush..."))
```

***Second Approach*** - using logic

```python
class Sol:
  def numOfSegments(self , s : str) -> int:
    s.split()
    count = 0

    for char in range(len(s)):
      if s[char] != "" and (char == 0 or s[char - 1] == " "):
        count = count + 1

    return count


s = Sol()
```

```python
print(s.numOfSegments(s = "Heyy , My name is Ayush..."))
```
