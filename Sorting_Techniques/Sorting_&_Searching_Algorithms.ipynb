# ***Bubble Sort***

### Working of Bubble Sort

Bubble Sort is a simple sorting algorithm that repeatedly steps through the list, compares adjacent elements and swaps them if they are in the wrong order. The pass through the list is repeated until no swaps are needed, which indicates that the list is sorted.

**How it works:**
1.  It starts by comparing the first two elements. If the first element is greater than the second, they are swapped.
2.  It then moves to the next pair of elements and repeats the comparison and swap.
3.  This process continues until the end of the list. After the first pass, the largest element will be at the end of the list.
4.  The algorithm then repeats the process for the remaining unsorted part of the list, excluding the element(s) already in their correct sorted position.
5.  This continues until no swaps are performed in a pass, indicating the list is sorted.

# ***Bubble Sort Algorithm***

```
procedure bubbleSort( list : array of items )

  loop = list.count;

  for i = 0 to loop-1 do:
    for j = 0 to loop-1 do:
      if list[j] > list[j+1] then
        swap( list[j], list[j+1] )
      end if
    end for
  end for

end procedure
```

### Time and Space Complexity of Bubble Sort

**Time Complexity:**

*   **Worst Case: O(n^2)**
    This occurs when the array is in reverse sorted order. For every element, the algorithm has to compare it with almost every other element, leading to n-1 passes, and in each pass, approximately n comparisons.
*   **Average Case: O(n^2)**
    For a randomly ordered array, the average case time complexity remains O(n^2) because, on average, a significant number of swaps and comparisons are still required.
*   **Best Case: O(n)**
    This occurs when the array is already sorted. The algorithm will still make one full pass to check if any swaps are needed. If no swaps occur, it realizes the list is sorted and terminates early. Modern bubble sort implementations often include a flag to detect if any swaps were made in a pass, allowing for this O(n) best-case performance.

**Space Complexity:**

*   **O(1)**
    Bubble sort is an in-place sorting algorithm, meaning it only requires a constant amount of additional memory space, regardless of the input size (n). It uses only a few temporary variables for comparisons and swaps.

***For Ascending Order***

````python
from typing import List

class Sol:
  def sortArray(self , nums : List[int]) -> List[int]:
    n = len(nums)

    for i in range(0 , n):
      for j in range(0 , n - i - 1):
        if nums[j] > nums[j + 1]:   # condition for ascending order
          nums[j] , nums[j + 1] = nums[j + 1] , nums[j]

    return nums

s = Sol()
print(s.sortArray(nums = [3 , 2 , 4 , 1 , 6 , 5]))
````

***For Descending Order***

````python
from typing import List

class Sol:
  def sortArray(self , nums : List[int]) -> List[int]:
    n = len(nums)

    for i in range(0 , n):
      for j in range(0 , n - i - 1):
        if nums[j] < nums[j + 1]:
          nums[j] , nums[j + 1] = nums[j + 1] , nums[j]

    return nums
s = Sol()
print(s.sortArray(nums = [3 , 2 , 4 , 1 , 6 , 5]))
````

# ***Insertion Sort***

### Working of Insertion Sort

Insertion sort is a simple sorting algorithm that builds the final sorted array (or list) one item at a time. It is much less efficient on large lists than more advanced algorithms such as quicksort, heapsort, or merge sort. However, it is an efficient algorithm for small arrays.

**How it works:**
1.  It iterates through the input array and removes one element per iteration to find the correct position and inserts it there.
2.  The comparison process starts from the end of the already sorted part of the array (which initially contains only the first element).
3.  Elements are shifted one position to the right until a suitable position is found for the element being inserted.
4.  The element is then inserted into its correct position.
5.  This process is repeated until all elements have been inserted into the sorted part of the array.

### Insertion Sort Algorithm

```
procedure insertionSort( list : array of items )

  for i = 1 to list.count - 1 do:
    key = list[i]
    j = i - 1

    while j >= 0 and list[j] > key do:
      list[j + 1] = list[j]
      j = j - 1
    end while

    list[j + 1] = key
  end for

end procedure
```

### Time and Space Complexity of Insertion Sort

**Time Complexity:**

*   **Worst Case: O(n^2)**
    This occurs when the array is in reverse sorted order. Each element must be compared with and shifted past every already sorted element.
*   **Average Case: O(n^2)**
    For a randomly ordered array, the average number of comparisons and shifts is proportional to n^2.
*   **Best Case: O(n)**
    This occurs when the array is already sorted. The inner while loop never executes, and the algorithm simply iterates through the array once, performing n-1 comparisons.

**Space Complexity:**

*   **O(1)**
    Insertion sort is an in-place sorting algorithm, requiring only a constant amount of additional memory space (for `key` and `j` variables), regardless of the input size (n).

### Insertion Sort Code Hinglish Explanation

Yeh code Insertion Sort algorithm ko implement karta hai. Chalo dekhte hain kaise kaam karta hai:

1.  **`class Sol:`**
    Yeh ek class hai jiske andar hum apna sorting function define karenge.

2.  **`def sortArray(self, nums: List[int]) -> List[int]:`**
    Yeh main function hai jo ek list `nums` input leta hai (integers ki list) aur usko sort karke return karta hai.

3.  **`n = len(nums)`**
    Hum list ki length `n` variable mein store kar lete hain.

4.  **`for i in range(1, n):`**
    Yeh outer loop `i` ko 1 se leke `n-1` tak chalata hai. Hum `i=1` se start karte hain kyunki hum maan kar chalte hain ki pehla element (index 0) already sorted hai.

5.  **`key = nums[i]`**
    Har iteration mein, `nums[i]` ko `key` variable mein store karte hain. `key` woh element hai jise hum sahi position par insert karna chahte hain sorted part mein.

6.  **`j = i - 1`**
    `j` pointer ko `i-1` par set karte hain. `j` sorted part ka last element point karta hai.

7.  **`while j >= 0 and nums[j] > key:`**
    Yeh inner loop chalta hai jab tak `j` 0 se bada ya barabar hai (matlab hum sorted part ke start tak nahi pahunche) aur `nums[j]` (sorted part ka current element) `key` se bada hai.

8.  **`nums[j + 1] = nums[j]`**
    Agar `nums[j]` `key` se bada hai, toh hum `nums[j]` ko ek position right shift kar dete hain (`nums[j+1]` par).

9.  **`j = j - 1`**
    `j` ko ek step left move karte hain, taaki hum sorted part ke agle element ko check kar sakein.

10. **`nums[j + 1] = key`**
    Jab inner loop khatam hota hai (ya toh `j` negative ho gaya, ya `nums[j]` `key` se chhota ya barabar hai), toh `key` ko uski sahi position (`j+1`) par insert kar dete hain.

11. **`return nums`**
    Jab outer loop poora ho jata hai, toh fully sorted list `nums` ko return kar dete hain.

12. **`s = Sol()`**
    `Sol` class ka ek object `s` banate hain.

13. **`print(s.sortArray(nums = [3, 2, 4, 1, 6, 5]))`**
    `sortArray` function ko call karte hain `[3, 2, 4, 1, 6, 5]` list ke saath aur sorted result print karte hain. Output `[1, 2, 3, 4, 5, 6]` aayega.

***Ascending Order***

````python
from typing import List

class Sol:
  def sortArray(self , nums : List[int]) -> List[int]:
    n = len(nums)

    for i in range(1 , n):
      key = nums[i]  # key = 4
      j = i - 1      # j = 2 - 1 = 1

      # condition for ascending order if nums[j] > key
      while j >= 0 and nums[j] > key:
        nums[j + 1] = nums[j]
        j = j - 1

      nums[j + 1] = key

    return nums

s = Sol()
print(s.sortArray(nums = [3 , 2 , 4 , 1 , 6 , 5]))
````

***Descending Order***

````python
class Sol:
  def sortArray(self , nums : List[int]) -> List[int]:
    n = len(nums)

    for i in range(1 , n):
      key = nums[i]
      j = i - 1

      while j >= 0 and nums[j] < key:
        nums[j + 1] = nums[j]
        j = j - 1

      nums[j + 1] = key

    return nums

s = Sol()
print(s.sortArray(nums = [3 , 2 , 4 , 1 , 6 , 5]))
````

# ***Selection Sort***

### Working of Selection Sort

Selection Sort is a simple comparison-based sorting algorithm that divides the input list into two parts: a sorted sublist and an unsorted sublist. It repeatedly finds the minimum element from the unsorted sublist and moves it to the sorted sublist.

**How it works:**
1.  **Initialization:** The entire list is initially considered unsorted.
2.  **Iteration:** The algorithm iterates through the unsorted part of the list.
3.  **Find Minimum:** In each iteration, it finds the smallest (or largest, depending on the sort order) element in the unsorted sublist.
4.  **Swap:** It then swaps this found minimum element with the first element of the unsorted sublist. This effectively moves the minimum element to its correct sorted position.
5.  **Boundary Adjustment:** The boundary between the sorted and unsorted sublists shifts one position to the right.
6.  **Repeat:** Steps 2-5 are repeated until the entire list is sorted.

### Selection Sort Algorithm

```
procedure selectionSort( list : array of items )

  n = list.length

  for i = 0 to n - 2 do:
    min_idx = i

    for j = i + 1 to n - 1 do:
      if list[j] < list[min_idx] then
        min_idx = j
      end if
    end for

    // Swap the found minimum element with the first element of the unsorted part
    swap( list[i], list[min_idx] )
  end for

end procedure
```

### Time and Space Complexity of Selection Sort

**Time Complexity:**

*   **Worst Case: O(n^2)**
    This occurs regardless of the initial order of the array. The algorithm always performs `n-1` passes, and in each pass, it makes a comparison with approximately `n` elements to find the minimum. This results in a quadratic number of comparisons.
*   **Average Case: O(n^2)**
    Similar to the worst case, the average case time complexity remains O(n^2) because the number of comparisons is always the same, irrespective of the input data's arrangement.
*   **Best Case: O(n^2)**
    Even if the array is already sorted, Selection Sort still needs to traverse the entire unsorted portion to find the minimum element, resulting in O(n^2) comparisons. The number of swaps would be minimal (zero if already sorted), but comparisons dominate the complexity.

**Space Complexity:**

*   **O(1)**
    Selection Sort is an in-place sorting algorithm. It only requires a constant amount of additional memory space for temporary variables (like `min_idx` and for swapping elements), regardless of the input size (n).

***Ascending Order***

````python
from typing import List

class Solution:
  def selectionSort(self , nums : List[int]) -> List[int]:
    n = len(nums)
    for i in range(n):
      min_idx = i

      for j in range(i + 1 , n):
        if nums[j] < nums[min_idx]:
          min_idx = j

      nums[i] , nums[min_idx] = nums[min_idx] , nums[i]

    return nums

s = Solution()
print(s.selectionSort(nums = [3 , 2 , 4 , 1 , 6 , 5]))
````

***Descending Order***

````python
from typing import List

class Solution:
  def selectionSort(self , nums : List[int]) -> List[int]:
    n = len(nums)

    for i in range(n):
      min_idx = i

      for j in range(i + 1 , n):
        if nums[j] > nums[min_idx]:
          min_idx = j

      nums[i] , nums[min_idx] = nums[min_idx] , nums[i]

    return nums

s = Solution()
print(s.selectionSort(nums = [3 , 2 , 4 , 1 , 6 , 5]))
````

# ***Merge Sort***

### Working of Merge Sort

Merge Sort is a highly efficient, comparison-based sorting algorithm that follows the **divide and conquer** paradigm. It recursively divides an unsorted list into sublists until each sublist has only one element (which is considered sorted). Then, it repeatedly merges these sublists to produce new sorted sublists until there is only one sorted list remaining.

**How it works (Divide and Conquer):**

1.  **Divide:** The unsorted list is divided into `n` sublists, each containing one element. A list with one element is considered sorted.
2.  **Conquer (Merge):** Repeatedly merge sublists to produce new sorted sublists until there is only one sorted list remaining. This merging process compares elements from two sublists and places them in the correct order into a new, larger sorted sublist.

### Merge Sort Algorithm

```
procedure mergeSort( list )
  if list.length <= 1 then
    return list
  end if

  mid = list.length / 2
  left_half = list[0 to mid-1]
  right_half = list[mid to list.length-1]

  left_sorted = mergeSort(left_half)
  right_sorted = mergeSort(right_half)

  return merge(left_sorted, right_sorted)
end procedure

procedure merge( left_list, right_list )
  result = empty list
  i = 0  // pointer for left_list
  j = 0  // pointer for right_list

  while i < left_list.length and j < right_list.length do:
    if left_list[i] <= right_list[j] then
      add left_list[i] to result
      i = i + 1
    else
      add right_list[j] to result
      j = j + 1
    end if
  end while

  // Add any remaining elements
  while i < left_list.length do:
    add left_list[i] to result
    i = i + 1
  end while

  while j < right_list.length do:
    add right_list[j] to result
    j = j + 1
  end while

  return result
end procedure
```

### Time and Space Complexity of Merge Sort

**Time Complexity:**

*   **Worst Case: O(n log n)**
    Merge Sort always divides the array into two halves, and merging two sorted halves takes linear time (O(n)). The division creates `log n` levels of recursion, and each level involves O(n) work (merging). Thus, `n * log n`.
*   **Average Case: O(n log n)**
    The average case performance is also O(n log n) because the algorithm consistently performs the divide and conquer steps, regardless of the initial arrangement of elements.
*   **Best Case: O(n log n)**
    Even if the array is already sorted, Merge Sort will still go through the entire process of dividing and merging, resulting in O(n log n) time complexity. It doesn't have an O(n) best case like some other algorithms.

**Space Complexity:**

*   **O(n)**
    Merge Sort requires an auxiliary space proportional to the input size (n) to store the merged sublists during the merge operation. This is because temporary arrays are created to hold the elements being merged.

### Real-Life Example: Sorting a Deck of Cards

Imagine you have a very large deck of unsorted playing cards and you want to sort them by rank. Using Merge Sort, you would:

1.  **Divide:** Split the large deck into two smaller piles.
2.  **Recursively Divide:** Keep splitting each pile in half until you have many tiny piles, each with only one card (a single card is sorted).
3.  **Merge:** Start combining two tiny sorted piles into a slightly larger sorted pile by taking the lowest card from the top of each pile and placing it into the new pile.
4.  **Repeat:** Continue merging these larger sorted piles until you eventually merge the two largest sorted halves back into one fully sorted deck.

````python
from typing import List

class Solution:
  def conquer(self , nums , left , mid , right):
    a = []
    b = []

    for i in range(left , mid + 1):
      a.append(nums[i])
    for i in range(mid + 1 , right + 1):
      b.append(nums[i])

    i , j , k = 0 , 0 , left

    while k <= right:
      if i == len(a):
        nums[k] = b[j]
        j = j + 1
        k = k + 1

      elif j == len(b):
        nums[k] = a[i]
        i = i + 1
        k = k + 1

      elif a[i] < b[j]:
        nums[k] = a[i]
        i = i + 1
        k = k + 1

      else:
        nums[k] = b[j]
        j = j + 1
        k = k + 1

  def divide(self , nums , left , right):
    if left >= right:  # base case
      return

    mid = (left + right) // 2
    self.divide(nums , left , mid)
    self.divide(nums , mid + 1 , right)
    self.conquer(nums , left , mid , right)


  def mergeSort(self , nums : List[int]) -> List[int]:
    self.divide(nums , left = 0 , right = len(nums) - 1)

    return nums

s = Solution()
print(s.mergeSort(nums = [1 , 5 , 6 , 2 , 4 , 3]))
````

# ***Quick Sort***

### Working of Quick Sort

Quick Sort is an efficient, comparison-based sorting algorithm that uses a **divide and conquer** strategy. It picks an element as a 'pivot' and partitions the array around this pivot, such that all elements smaller than the pivot come before it, and all elements greater than the pivot come after it. This process is then applied recursively to the sub-arrays.

**How it works (in simple words):**
1.  **Pick a 'Leader' (Pivot):** You choose one item from your list and call it the 'pivot'. Think of it as a reference point.
2.  **Separate into Two Groups:** You then go through the rest of the items. If an item is smaller than your 'leader', you put it in a 'smaller' group. If it's larger, you put it in a 'larger' group.
3.  **Place the Leader:** Once you've sorted everything into 'smaller' and 'larger' groups, you place the 'leader' (pivot) right in between them. Now, the 'leader' is in its final sorted position.
4.  **Repeat for Groups:** You then take the 'smaller' group and the 'larger' group and do the exact same thing with them: pick a new 'leader' for each group, separate their items, and place their leaders.
5.  **Keep Going:** You keep doing this until each group is so small that it only has one item (which is already sorted). Then, when you put all these single-item groups back together, your entire list is sorted!

### Quick Sort Algorithm (Pseudo-code)

```
procedure quickSort( array, low, high )
  if low < high then
    pivot_index = partition( array, low, high )
    quickSort( array, low, pivot_index - 1 )
    quickSort( array, pivot_index + 1, high )
  end if
end procedure

procedure partition( array, low, high )
  pivot = array[high] // Choose the last element as pivot
  i = low - 1         // Index of smaller element

  for j = low to high - 1 do:
    if array[j] <= pivot then
      i = i + 1
      swap( array[i], array[j] )
    end if
  end for

  swap( array[i + 1], array[high] )
  return i + 1
end procedure
```

### Real-Life Example: Sorting Books on a Shelf

Imagine you have a messy shelf of books, and you want to sort them alphabetically. You can use Quick Sort:

1.  **Pick a Pivot:** You pick a book, say "Moby Dick", from the shelf. This is your pivot.
2.  **Partition:** You quickly scan through the other books. All books that start with letters A-L you put to the left of "Moby Dick". All books that start with N-Z you put to the right. "Moby Dick" is now in its correct alphabetical position relative to all other books.
3.  **Recursively Sort Sub-shelves:** Now you have two smaller, unsorted piles of books (one for A-L, one for N-Z). You repeat the process for each of these smaller piles: pick a new pivot book (e.g., "Harry Potter" for the A-L pile), partition those books, and so on.
4.  **Combine:** You keep doing this until each segment of the shelf is sorted, and eventually, the entire shelf is sorted.

### Time and Space Complexity of Quick Sort

**Time Complexity:**

*   **Worst Case: O(n^2)**
    This occurs when the pivot selection consistently leads to highly unbalanced partitions, such as always picking the smallest or largest element as the pivot in an already sorted or reverse-sorted array. Each partitioning step then takes O(n) time, and there are O(n) such steps, leading to O(n^2).

*   **Average Case: O(n log n)**
    On average, Quick Sort performs very well. When pivots divide the array into roughly equal halves, the recursion depth is O(log n), and each level involves O(n) work for partitioning. This yields an average time complexity of O(n log n).

*   **Best Case: O(n log n)**
    This occurs when the pivot always divides the array into two equal or nearly equal halves. The logic is similar to the average case, leading to O(n log n).

**Space Complexity:**

*   **Worst Case: O(n)**
    In the worst-case scenario (unbalanced partitions), the recursion stack depth can go up to O(n), requiring O(n) auxiliary space. This often happens with naive pivot selection on already sorted arrays.

*   **Average Case: O(log n)**
    On average, with good pivot selection, the recursion stack depth is O(log n), making the space complexity O(log n). This is because the algorithm only needs to store a logarithmic number of active function calls on the stack at any given time.

### Quick Sort Code (Ascending Order)

````python
from typing import List

class Solution:
  def quickSort(self , nums : List[int]) -> List[int]:
    if len(nums) <= 1:
      return nums

    pivot_element = nums[-1]
    left = [x for x in nums if x < pivot_element]
    mid = [x for x in nums if x == pivot_element]
    right = [x for x in nums if x > pivot_element]

    return self.quickSort(left) + mid + self.quickSort(right)


s = Solution()
print(s.quickSort(nums = [1 , 5 , 6 , 2 , 4 , 3]))
````

### Quick Sort Code (Descending Order)

````python
from typing import List

class Solution:
  def quickSort(self , nums : List[int]) -> List[int]:
    if len(nums) <= 1:
      return nums

    pivot_element = nums[-1]
    left = [x for x in nums if x > pivot_element]  # comparison signs flip in descending order for left and right both
    mid = [x for x in nums if x == pivot_element]
    right = [x for x in nums if x < pivot_element] # comparison signs flip in descending order for left and right both

    return self.quickSort(left) + mid + self.quickSort(right)


s = Solution()
print(s.quickSort(nums = [1 , 5 , 6 , 2 , 4 , 3]))
````

# ***Counting Sort***

### Working of Counting Sort

Counting Sort is a non-comparison based sorting algorithm that works by counting the number of occurrences of each distinct element in an array. It is highly efficient for sorting data within a specific range.

**How it works (in simple words):**
1.  **Find the Range:** First, you need to know the smallest and largest numbers in your list. This helps you determine the range of values you'll be dealing with.
2.  **Create a 'Counter' List:** You then create a new list (let's call it `count_array`) that's big enough to hold a count for every number in your range. Initially, all counts are zero.
3.  **Count Occurrences:** Go through your original unsorted list. For each number, increment its corresponding counter in the `count_array`. For example, if you see the number '3', you increase the count at index '3' in your `count_array`.
4.  **Cumulative Counts (Optional but helpful):** Transform the `count_array` so that each element at index `i` stores the actual position of `i` in the sorted output array. This is done by adding the current count to the previous count (`count_array[i] = count_array[i] + count_array[i-1]`).
5.  **Build the Sorted Array:** Create an empty output array of the same size as your original list. Now, iterate through the original unsorted list from right to left (or left to right, depending on the stable property requirement):
    *   For each number in the original list, look up its cumulative count in the `count_array`. This count tells you where that number should go in the sorted output array.
    *   Place the number in the `output_array` at that position.
    *   Decrement the count in the `count_array` for that number. This ensures that if there are duplicate numbers, the next occurrence of that number gets placed at the correct preceding position, making the sort stable.
6.  **Copy Back:** Finally, copy the elements from the `output_array` back into the original array (or use the output array as the final sorted result).

### Counting Sort Algorithm (Pseudo-code)

```
procedure countingSort( array )
  n = array.length
  output_array = new Array(n) // Result array

  // Find the maximum element to determine the range
  max_val = array[0]
  for i = 1 to n - 1 do:
    if array[i] > max_val then
      max_val = array[i]
    end if
  end for

  // Create count_array and initialize with zeros
  count_array = new Array(max_val + 1)
  for i = 0 to max_val do:
    count_array[i] = 0
  end for

  // Store the count of each element
  for i = 0 to n - 1 do:
    count_array[array[i]] = count_array[array[i]] + 1
  end for

  // Store cumulative count
  for i = 1 to max_val do:
    count_array[i] = count_array[i] + count_array[i - 1]
  end for

  // Place elements into the output_array from right to left
  // This ensures stability for duplicate elements
  for i = n - 1 down to 0 do:
    output_array[count_array[array[i]] - 1] = array[i]
    count_array[array[i]] = count_array[array[i]] - 1
  end for

  // Copy the sorted elements back to the original array
  for i = 0 to n - 1 do:
    array[i] = output_array[i]
  end for

  return array
end procedure
```

### Dry Run Example: Sorting `[1, 4, 1, 2, 7, 5, 2]`

Let's sort the array `A = [1, 4, 1, 2, 7, 5, 2]` using Counting Sort.

**Step 1: Find the maximum element.**
*   `max_val` in `A` is 7.

**Step 2: Initialize `count_array` with zeros.**
*   Since `max_val` is 7, `count_array` will have size `7 + 1 = 8` (indices 0 to 7).
*   `count_array = [0, 0, 0, 0, 0, 0, 0, 0]`

**Step 3: Store the count of each element.**
*   Iterate through `A`:
    *   `1`: `count_array[1]` becomes 1.
    *   `4`: `count_array[4]` becomes 1.
    *   `1`: `count_array[1]` becomes 2.
    *   `2`: `count_array[2]` becomes 1.
    *   `7`: `count_array[7]` becomes 1.
    *   `5`: `count_array[5]` becomes 1.
    *   `2`: `count_array[2]` becomes 2.
*   After this step:
    *   `count_array = [0, 2, 2, 0, 1, 1, 0, 1]` (Index 0 is always 0 as we assume non-negative integers)

**Step 4: Store cumulative count.**
*   `count_array[0]` remains 0.
*   `count_array[1] = count_array[1] + count_array[0] = 2 + 0 = 2`
*   `count_array[2] = count_array[2] + count_array[1] = 2 + 2 = 4`
*   `count_array[3] = count_array[3] + count_array[2] = 0 + 4 = 4`
*   `count_array[4] = count_array[4] + count_array[3] = 1 + 4 = 5`
*   `count_array[5] = count_array[5] + count_array[4] = 1 + 5 = 6`
*   `count_array[6] = count_array[6] + count_array[5] = 0 + 6 = 6`
*   `count_array[7] = count_array[7] + count_array[6] = 1 + 6 = 7`
*   After this step:
    *   `count_array = [0, 2, 4, 4, 5, 6, 6, 7]`
    *   This array now tells us the final position (plus one) of each element. For example, the last '1' will go at index `count_array[1] - 1 = 2 - 1 = 1`. The last '2' will go at index `count_array[2] - 1 = 4 - 1 = 3`.

**Step 5: Place elements into `output_array` (iterate `A` from right to left).**
*   `output_array = [_, _, _, _, _, _, _]` (size 7)

| Current Element `A[i]` | `count_array[A[i]]` | `output_array` Update | `count_array` Update |
| :--------------------- | :------------------ | :-------------------- | :------------------- |
| `A[6] = 2`             | `count_array[2] = 4` | `output_array[3] = 2` | `count_array[2] = 3` |

| `A[0] = 1`             | `count_array[1] = 1` | `output_array[0] = 1` | `count_array[1] = 0` |

*   Final `output_array = [1, 1, 2, 2, 4, 5, 7]`

**Step 6: Copy `output_array` back to `A` (if required).**

### Time and Space Complexity of Counting Sort

**Time Complexity:**

*   **O(n + k)**
    *   `n` is the number of elements in the input array.
    *   `k` is the range of the input numbers (i.e., `max_val - min_val + 1`).
    *   **Finding `max_val`:** O(n) to iterate through the array.
    *   **Initializing `count_array`:** O(k).
    *   **Counting occurrences:** O(n) to iterate through the array.
    *   **Calculating cumulative counts:** O(k).
    *   **Building the output array:** O(n) to iterate through the original array and place elements.
    *   **Copying back (optional):** O(n).
    *   Therefore, the total time complexity is O(n + k).

**Space Complexity:**

*   **O(n + k)**
    *   **`count_array`:** Requires O(k) space to store the counts of each element.
    *   **`output_array`:** Requires O(n) space to store the sorted elements.
    *   Therefore, the total space complexity is O(n + k).

**Important Considerations:**
*   Counting Sort is most efficient when the range of input data (`k`) is not significantly larger than the number of elements (`n`). If `k` is very large, the space and time complexity become dominated by `k`, making it less efficient than comparison-based sorts like Merge Sort or Quick Sort (which have O(n log n) time complexity).
*   It is a **stable** sorting algorithm, meaning that elements with identical values maintain their relative order in the sorted output.

````python
# example - arr = [2 , 1 , 0 , 0 , 2 , 3 , 0 , 5 , 4 , 4]

from typing import List

class Solution:
  def countingSort(self , arr : List[int]) -> List[int]:
    n = len(arr)  # n = 10
    max_element = max(arr)   # max_element = 5

    freq = [0] * (max_element + 1)   # we take an empty array named freq and it's length is max_element + 1
    # why we make this array , to store the frequency of elements of arr
    # initially freq = [0 , 0 , 0 , 0 , 0 , 0]

    for i in arr:
      freq[i] = freq[i] + 1
    # then freq = [3 , 1 , 2 , 1 , 2 , 1]

    nums = []  # I make an empty list because , I wanna store , numbers of freq as many time they are present in freq ,

    for i in range(0 , max_element + 1):
      while freq[i] > 0:
        nums.append(i)
        freq[i] = freq[i] - 1
    # nums = [0 , 0 , 0 , 1 , 2 , 2 , 3 , 4 , 4 , 5] becoz 0 present 3 times , 1 present 1 time , 2 present only 2 times , and so on...

    # doing this we got a sorted array

    return nums

s = Solution()
print(s.countingSort(arr = [2 , 1 , 0 , 0 , 2 , 3 , 0 , 5 , 4 , 4]))
````

# ***Searching Algorithms***

### ***Linear Search***

**Working of Linear Search**

Linear search (also known as sequential search) is a very simple search algorithm that sequentially checks each element of the list until a match is found or the whole list has been searched. It is the most straightforward searching method.

**How it works:**
1.  Start from the first element of the list and compare it with the target value.
2.  If the element matches the target value, the search is successful, and its index is returned.
3.  If the element does not match, move to the next element and repeat the comparison.
4.  Continue this process until either the target value is found or the end of the list is reached.
5.  If the entire list is traversed and the target value is not found, the search is unsuccessful, and typically a special value (like -1) is returned.

**Time Complexity:**

*   **Worst Case: O(n)**
    This occurs when the target element is at the end of the list or not present in the list. The algorithm has to iterate through all 'n' elements.
*   **Average Case: O(n)**
    On average, if the element is present, it will be found roughly in the middle of the list, leading to O(n) comparisons.
*   **Best Case: O(1)**
    This occurs when the target element is the first element of the list. The algorithm finds the element in a single comparison.

**Space Complexity:**

*   **O(1)**
    Linear search is an in-place algorithm as it does not require any additional storage space proportional to the input size 'n'. It only uses a few temporary variables for iteration and comparison.

````python
from typing import List

class Solution:
  def linearSearch(self , nums : List[int] , target : int):
    n = len(nums)
    res = []

    if not nums:
      return f"Target not found"

    for i in nums:
      if i == target:
        res.append(nums.index(i))   # we store indices in an array , in case if their are multiple indices containing target.

    return f"Target found at following indices : {res}"

s = Solution()
print(s.linearSearch(nums = [5 , 1 , 8 , 2 , 6 , 9 , 4] , target = 2))
````

### ***Binary Search***

Binary search is an efficient algorithm for finding an item from a **sorted** list of items. It works by repeatedly dividing in half the portion of the list that could contain the item, until you've narrowed down the possible locations to just one. Binary search is significantly faster than linear search for large sorted arrays.

**Working of Binary Search**

1.  **Start with the middle element:** Compare the target value with the middle element of the sorted array.
2.  **Match Found:** If the target value is equal to the middle element, the search is successful, and its index is returned.
3.  **Adjust search range:**
    *   If the target value is *less than* the middle element, then the target must lie in the left half of the array. Discard the right half and repeat the search on the left half.
    *   If the target value is *greater than* the middle element, then the target must lie in the right half of the array. Discard the left half and repeat the search on the right half.
4.  **Repeat:** Continue steps 1-3 until the target is found or the search range becomes empty (meaning the target is not in the array).

**Time Complexity:**

*   **Worst Case: O(log n)**
    In the worst case, the algorithm halves the search space in each step. This logarithmic behavior occurs because each comparison effectively eliminates half of the remaining elements from consideration.
*   **Average Case: O(log n)**
    On average, Binary Search performs a logarithmic number of comparisons, similar to the worst case.
*   **Best Case: O(1)**
    This occurs when the target element is the middle element of the array on the first comparison.

**Space Complexity:**

*   **Iterative Implementation: O(1)**
    The iterative version of Binary Search uses a constant amount of extra space, regardless of the input size, as it only requires a few variables to store pointers (low, high, mid).
*   **Recursive Implementation: O(log n)**
    The recursive version uses space on the call stack. In the worst case, the recursion depth is `log n`, leading to O(log n) space complexity.

````python
from typing import List

class Solution:
  def binarySearch(self , nums : List[int] , target):
    n = len(nums)

    left = 0
    right = n - 1

    while left <= right:
      middle = (left + right) // 2
      if target == nums[middle]:
        return f"Target found at index : {middle}"

      elif nums[middle] <= target:
        left = middle + 1

      else:
        right = middle


    return f"Target not found"

s = Solution()
print(s.binarySearch(nums = [1 , 2  , 5 , 6 , 7 , 9] , target = 2))
````

````python
# Lower Bound - index of first element in a sorted array that is greater than or equal to target value.
from typing import List

class Solution:
  def lowerBound(self , nums : List[int] , target : int):
    n = len(nums)
    l = 0
    r = n - 1
    ans = n     # we let that lower bound already last index (n)

    while l <= r:
      mid = (l + r) // 2

      if nums[mid] >= target:   # if our number is greater than or equal to our target , then that is our answer
        ans = mid
        r = mid - 1   # but I did r = mid - 1 , becoz I want the smalled index whose value is greater than or equal to target.

      else:
        l = mid + 1   # else I increase / shift left side to (mid + 1)

    return ans

s = Solution()
print(s.lowerBound(nums = [2 , 3  , 8 , 12 , 20 , 23 , 40] , target = 12))
````

````python
# Upper Bound - index of first element in a sorted array that is greater than target value.

from typing import List

class Solution:
  def upperBound(self , nums : List[int] , target : int):
    n = len(nums)
    l = 0
    r = n - 1
    ans = n

    while l <= r:
      mid = (l + r) // 2

      if nums[mid] > target:   # index of first element in a sorted array that is greater than target value.
        ans = mid   # if yes , then ans = mid index
        r = mid - 1  # but then too I check by doing l = mid + 1 , if any number exist that satisfies the condition.

      else:
        l = mid + 1  #

    return ans

s = Solution()
print(s.upperBound(nums = [3 , 3 , 5 , 5 , 5 , 5  , 6 , 6 , 9 , 10 , 10] , target = 5))
````

````python
# return square root of x without using built-in function.

class Solution:
  def sqrt(self , x : int) -> int:
    left = 0
    right = x

    while left <= right:
      mid = (left + right) // 2

      if mid * mid == x:
        return mid

      elif mid * mid < x:
        left = mid + 1

      else:
        right = mid - 1

    return right

s = Solution()
print(s.sqrt(x = 18))
print(s.sqrt(x = 25))
````

````python
from typing import List

class Solution:
  def searchMatrix(self , matrix : List[List[int]] , target : int) -> bool:
    rows = len(matrix)
    cols = len(matrix[0])

    l = 0
    r = (rows * cols) - 1

    while l <= r:
      mid = (l + r) // 2   # mid = (0 + 11) // 2 = 5
      # but , the twist is how will I decide the real indexing , seeing the above image , I will assign some imaginary indices to each element of matrix starting 0 to 11 simply.
      # then I calulate mid , then then then , to find the real index of that particular element , see this : mid // cols = row number & mid % cols = column number.

      row_pos = mid // cols   # 5 // 4 = 1
      col_pos = mid % cols    # 5 % 4 = 1

      if matrix[row_pos][col_pos] == target:
        return True

      elif matrix[row_pos][col_pos] > target:
        r = mid - 1

      else:
        l = mid + 1

    return False


s = Solution()
print(s.searchMatrix([[1,3,5,7] , [10,11,16,20] , [23,30,34,60]] , target = 11))
````
