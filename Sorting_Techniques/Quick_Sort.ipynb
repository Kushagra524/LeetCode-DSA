Quick Sort is an efficient, comparison-based sorting algorithm that works on the principle of a **divide-and-conquer** strategy. Here's a detailed and simple breakdown:

### The Core Idea: Divide and Conquer
Imagine you have a messy pile of books and you want to sort them by author. Instead of sorting the whole pile at once, Quick Sort suggests:

1.  **Pick a "pivot" book**: Choose one book from the pile. Let's say you pick a book by 'C.S. Lewis'. This book will be your pivot.
2.  **Partition the pile**: Now, go through the rest of the books. Put all books by authors *before* 'C.S. Lewis' into a pile on the left. Put all books by authors *after* 'C.S. Lewis' into a pile on the right. 'C.S. Lewis' stays in the middle.
3.  **Sort the smaller piles**: Now you have two smaller, unsorted piles (left and right). You recursively apply the same process to each of these smaller piles: pick a pivot, partition, and sort the sub-piles.
4.  **Combine (no work needed!)**: Once all the smaller piles are sorted, because of the way you partitioned them, the entire collection of books will be sorted! The "conquer" step is almost free because the partitioning places elements in their correct relative positions.

### How it Works (Algorithm Steps):

1.  **Choose a Pivot**: Select an element from the array to be the pivot. Common choices include:
    *   The first element.
    *   The last element.
    *   The middle element.
    *   A random element (often preferred to avoid worst-case scenarios).

2.  **Partitioning**: Rearrange the array such that all elements smaller than the pivot come before it, and all elements greater than the pivot come after it. Elements equal to the pivot can go on either side. After this step, the pivot element is in its final sorted position.

    *   **Example**: Array `[7, 2, 1, 6, 8, 5, 3, 4]`. Let's pick `4` (last element) as the pivot.
    *   After partitioning, the array might look like `[2, 1, 3, 4, 8, 5, 6, 7]`. Notice `4` is in its correct place, and everything to its left is smaller, everything to its right is larger.

3.  **Recursion**: Recursively apply the above two steps to the sub-array of elements with smaller values and the sub-array of elements with greater values. This continues until the sub-arrays have zero or one element, as single-element arrays are inherently sorted.

### Simple Code Snippet (Conceptual Python):

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    else:
        pivot = arr[len(arr) // 2] # Choose middle element as pivot
        left = [x for x in arr if x < pivot]
        middle = [x for x in arr if x == pivot]
        right = [x for x in arr if x > pivot]
        return quick_sort(left) + middle + quick_sort(right)

# Example usage:
my_list = [10, 7, 8, 9, 1, 5]
sorted_list = quick_sort(my_list)
print(sorted_list) # Output: [1, 5, 7, 8, 9, 10]
```

### Why is it popular?

*   **Efficiency**: On average, Quick Sort is very fast. Its average time complexity is `O(n log n)`, which means it scales well with larger datasets.
*   **In-place (mostly)**: Many implementations perform the sort "in-place," meaning they require minimal additional memory, which is efficient for memory usage.

### Things to Consider:

*   **Worst-Case**: If the pivot selection is consistently poor (e.g., always picking the smallest or largest element in an already sorted array), Quick Sort's performance can degrade to `O(n^2)`. Random pivot selection helps mitigate this.
*   **Stability**: Quick Sort is generally not a stable sorting algorithm, meaning the relative order of equal elements might change after sorting.

````python
from typing import List

class Solution:
  def quickSort(self , arr : List[int]) -> List[int]:
    if len(arr) <= 1:
      return arr

    pivot = arr[-1]
    left = [x for x in arr if x < pivot]
    mid = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return self.quickSort(left) + mid + self.quickSort(right)

s = Solution()
print(s.quickSort(arr = [10, 7, 8, 9, 1, 5]))
````
