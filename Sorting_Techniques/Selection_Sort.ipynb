# ***Selection Sort***

Selection Sort is a simple comparison-based sorting algorithm. It works by repeatedly finding the minimum element from the unsorted part of the list and putting it at the beginning of the sorted part.

Here's how it generally works:

1.  **Iterate through the array**: Start from the first element and iterate through the entire array.

2.  **Find the minimum element**: In each iteration, assume the current element is the minimum. Then, search for the actual minimum element in the *unsorted* part of the array (from the current position to the end).

3.  **Swap**: If a smaller element is found, swap it with the element at the current position. This moves the minimum element to its correct sorted position.

4.  **Repeat**: Move to the next position and repeat the process until the entire array is sorted.

**Example:**

Let's sort the array `[64, 25, 12, 22, 11]` using Selection Sort.

*   **First Pass:**
    *   Current position: Index 0 (value 64)
    *   Find minimum in `[64, 25, 12, 22, 11]`. The minimum is 11 at index 4.
    *   Swap 64 and 11. Array becomes `[11, 25, 12, 22, 64]`.

*   **Second Pass:**
    *   Current position: Index 1 (value 25)
    *   Find minimum in `[25, 12, 22, 64]` (the unsorted part).
    *   The minimum is 12 at index 2.
    *   Swap 25 and 12. Array becomes `[11, 12, 25, 22, 64]`.

*   **Third Pass:**
    *   Current position: Index 2 (value 25)
    *   Find minimum in `[25, 22, 64]`.
    *   The minimum is 22 at index 3.
    *   Swap 25 and 22. Array becomes `[11, 12, 22, 25, 64]`.

*   **Fourth Pass:**
    *   Current position: Index 3 (value 25)
    *   Find minimum in `[25, 64]`.
    *   The minimum is 25 at index 3 (already in place).
    *   No swap needed (or swap with itself). Array remains `[11, 12, 22, 25, 64]`.

The array is now sorted.

**Time Complexity:**

Selection Sort has a time complexity of O(n²) in all cases (best, average, and worst) because it always makes n-1 passes and, in each pass, it performs comparisons to find the minimum element in the remaining unsorted part.

```python
class Solution:
    def sortArray(self, nums: list[int]) -> list[int]:
        n = len(nums)
        for i in range(n):
            min_idx = i
            for j in range(i + 1, n):
                if nums[j] < nums[min_idx]:
                    min_idx = j
            nums[i], nums[min_idx] = nums[min_idx], nums[i]
        return nums

solver = Solution()
```

```python
print(solver.sortArray(nums = [64 , 25 , 12 , 22 , 11]))
```

```python
print(solver.sortArray(nums = [5, 1, 4, 2, 8]))
```

```python
print(solver.sortArray(nums = [1, 2, 3, 4, 5]))  # already sorted array
```
