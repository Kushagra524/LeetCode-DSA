# ***Insertion Sort*** -


arr = 3 , 2 , 8 , 1 , 7 , 6 , 4
      

```python
from typing import List

class Solution:
  def insertionSort(self , arr : List[int]) -> List[int]:
    n = len(arr)

    for i in range(1 , n):
      key = arr[i]
      j = i - 1

      while j >= 0 and key < arr[j]:
        arr[j + 1] = arr[j]
        j = j - 1

      arr[j + 1] = key
    return arr

s = Solution()
```

```python
print(s.insertionSort([40 , 20 , 30 , 10]))
```

***For desceding order***

```python
from typing import List

class Solution:
  def insertionSort(self, arr: List[int]) -> List[int]:
    n = len(arr)

    for i in range(1, n):
      key = arr[i]
      j = i - 1

      while j >= 0 and arr[j] < key:    # for descending order
        arr[j + 1] = arr[j]
        j -= 1

      arr[j + 1] = key

    return arr


s = Solution()
```

```python
print(s.insertionSort([40, 20, 30, 10]))
```

### Time Complexity Analysis of Insertion Sort

Insertion Sort's time complexity depends on the initial order of the input array.

*   **Best Case: O(n)**
    *   Occurs when the array is already sorted.
    *   In each pass of the outer loop, the inner `while` loop condition `key < arr[j]` is immediately false for most elements (or `j >= 0` becomes false after one comparison), meaning no swaps are needed or only one comparison occurs.
    *   The algorithm still iterates through all `n` elements, but the number of comparisons and shifts is minimal, leading to linear time complexity.

*   **Average Case: O(n^2)**
    *   Occurs when the elements are in a jumbled or random order.
    *   On average, for each element `arr[i]`, the inner `while` loop will perform roughly `i/2` comparisons and `i/2` shifts.
    *   Summing this up for `i` from `1` to `n-1` gives `1 + 2 + ... + (n-1)` comparisons/shifts, which is approximately `n^2/4`, hence O(n^2).

*   **Worst Case: O(n^2)**
    *   Occurs when the array is sorted in reverse order.
    *   For every element `arr[i]`, the inner `while` loop has to compare `key` with all `i` preceding elements and shift them one position to the right.
    *   This leads to `1 + 2 + ... + (n-1)` comparisons and shifts, which sums up to `n * (n-1) / 2` operations, resulting in quadratic time complexity O(n^2).

### Space Complexity

*   **Space Complexity: O(1)**
    *   Insertion Sort is an in-place sorting algorithm, meaning it sorts the array within the original array without requiring significant additional memory space.
    *   It only uses a constant amount of extra space for variables like `key` and `j`.

```python
# Question - sort this array using insertion sort

from typing import List

class Solution:
  def insertionSort(self , arr : List[int]) -> List[int]:
    n = len(arr)

    for i in range(1 , n):  # i = 3
      key = arr[i]    # key = 30
      j = i - 1   # j = 2

      while j >= 0 and key < arr[j]:  # 1 >= 0 and 30 <= 20
        arr[j + 1] = arr[j]   # arr[3] = arr[2]
        j = j - 1   # j = 1

      arr[j + 1] = key   # arr[2] = 30

    return arr

s = Solution()

# 10 , 20 , 30 , 40
```
