# Merge Sort Algorithm

Merge Sort is a **Divide and Conquer** algorithm. It works by recursively breaking down a list into several sub-lists until each sub-list consists of a single element and merging those sub-lists in a manner that results into a sorted list.

### How it Works:
1. **Divide:** Find the middle point to divide the array into two halves:
   `middle = left + (right - left) / 2`
2. **Conquer:** Recursively call `mergeSort` for the left half and the right half.
3. **Combine:** Merge the two halves sorted in step 2.

### Complexity:
*   **Time Complexity:** $O(n \log n)$ for all cases (best, average, and worst) because the array is always divided into two halves and takes linear time to merge two halves.
*   **Space Complexity:** $O(n)$ as it requires additional space for the temporary subarrays.

### Advantages:
*   Stable sort (preserves the relative order of equal elements).
*   Guaranteed $O(n \log n)$ performance.
*   Excellent for sorting linked lists.

***First approach***

```python
# arr = 1 5 6 2 3 4

from typing import List

class Solution:
  def conquer(self , nums , left , mid , right):  # this functions merges the divided array such that they result in a sorted array
    # now the logic is , elements from (left till mid) goes into arr1 and elements from mid + 1 right goes into arr2
    arr1 = []
    arr2 = []
    for i in range(left , mid + 1):  # arr1 =  1 , 5 , 6
      arr1.append(nums[i])   # we get elements in arr1 , but not sorted

    for i in range(mid + 1 , right + 1):  # arr2 = 2 , 3 , 4
      arr2.append(nums[i])   # we got elements in arr2 , but not sorted
    # Now I want to get elements from arr1 and arr2 together into a single array and sorted form

    i , j , k = 0 , 0 , left
    # i is the index of arr1
    # j is the index of arr2
    # left is the start points of our (nums)

    while k <= right:  # right = last index of nums
      if j == len(arr2):    # this means j has reached to end of arr2
        nums[k] = arr1[i]
        i = i + 1
        k = k + 1

      elif i == len(arr1):  # this means i has reached to end of arr1
        nums[k] = arr2[j]
        j = j + 1
        k = k + 1

      elif arr1[i] > arr2[j]: # Changed from < to > for descending order
        nums[k] = arr1[i]
        k = k + 1
        i = i + 1

      else:                # else condition if arr1[i] < arr2[j]
        nums[k] = arr2[j]
        j = j + 1
        k = k + 1

  def divide(self , nums , left , right):
    # this is my base case
    if left >= right:  # mean only when their is single elment left in the array
      return  # when left reaches right or greter than right , we simply return

    # recursive case
    mid = (left + right) // 2
    self.divide(nums , left , mid)  # we first divide left to mid
    self.divide(nums , mid + 1 , right)  # then we divide from (mid + 1) to right

    self.conquer(nums , left , mid , right)  # after our array is divided fully into individual elements , we conquer (merge) it.

  def sortArray(self , nums : List[int]) -> List[int]:
    self.divide(nums , left = 0 , right = len(nums) - 1)

    return nums

s = Solution()
```

```python
print(s.sortArray(nums = [5 , 2 , 3 , 4 , 1]))
```

### Descending Order Implementation

```python
from typing import List

class SolutionDescending:
  def conquer(self , nums , left , mid , right):
    arr1 = []
    arr2 = []
    for i in range(left , mid + 1):
      arr1.append(nums[i])

    for i in range(mid + 1 , right + 1):
      arr2.append(nums[i])

    i , j , k = 0 , 0 , left

    while k <= right:
      if j == len(arr2):
        nums[k] = arr1[i]
        i = i + 1
        k = k + 1

      elif i == len(arr1):
        nums[k] = arr2[j]
        j = j + 1
        k = k + 1

      elif arr1[i] > arr2[j]: # Comparison for descending order
        nums[k] = arr1[i]
        k = k + 1
        i = i + 1

      else:
        nums[k] = arr2[j]
        j = j + 1
        k = k + 1

  def divide(self , nums , left , right):
    if left >= right:
      return

    mid = (left + right) // 2
    self.divide(nums , left , mid)
    self.divide(nums , mid + 1 , right)
    self.conquer(nums , left , mid , right)

  def sortArray(self , nums : List[int]) -> List[int]:
    self.divide(nums , left = 0 , right = len(nums) - 1)
    return nums

desc_solver = SolutionDescending()
print("Descending sort result:", desc_solver.sortArray(nums = [5 , 2 , 3 , 4 , 1]))
```
