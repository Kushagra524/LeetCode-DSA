## ***The linearSearch function works by iterating through each element of the input list nums one by one. Here's a breakdown:***

1. found_indices = [ ]: An empty list called found_indices is initialized. This list will store the indices of all elements in nums that match the target.

2. for i in range(len(nums)): This loop iterates from the first element (index 0) to the last element of the nums list. The variable i represents the current index.

3. if nums[i] == target:: Inside the loop, at each iteration, it checks if the element at the current index i (i.e., nums[i]) is equal to the target value.

4. found_indices.append(i): If the element nums[i] matches the target, its index i is added to the found_indices list.

5. return found_indices: After checking all elements in the nums list, the function returns the found_indices list, which contains all the positions where the target was found. If the target was not found anywhere, the list will be empty.

```python
from typing import List

class Solution:
  def linearSearch(self , arr : List[int] , target : int) -> List[int]:    # this function returns a List of indices in the input array that match our particular Target

    found = []

    for num in range(0 , len(arr)):
      if arr[num] == target:
        found.append(f"index {num}")

    if len(found) > 0:
      return f"Target found at following indices : {found}"
    else:
      return f"Target not found"

sol = Solution()
```

```python
print(sol.linearSearch(arr = [1, 5, 2, 8, 5, 9, 5] , target = 5))
```

```python
print(sol.linearSearch(arr = [10, 20, 30, 40, 50] , target = 30))
```

```python
print(sol.linearSearch(arr = [1, 2, 3, 4, 5] , target = 9))
```
