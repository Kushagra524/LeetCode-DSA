## Binary Search Explanation

Binary search is an efficient algorithm for finding an item from a **sorted** list of items. It works by repeatedly dividing in half the portion of the list that could contain the item, until you've narrowed down the possible locations to just one.

Here's how it works step-by-step:

1.  **Start with the middle element**: Look at the element in the middle of the sorted list.
2.  **Compare**: Compare the target value you're looking for with the middle element.
    *   If the target value is equal to the middle element, you've found it!
    *   If the target value is less than the middle element, you know the target must be in the left half of the list.
    *   If the target value is greater than the middle element, you know the target must be in the right half of the list.
3.  **Repeat**: Discard the half of the list where the target cannot be, and repeat the process on the remaining half. This continues until the target element is found or the remaining portion of the list is empty (meaning the element is not in the list).

**Key Requirements:**
*   The list **must be sorted** for binary search to work correctly.

**Time Complexity:**
*   **O(log n)**: Binary search is very fast because it eliminates half of the remaining elements in each step, making it much more efficient than linear search (O(n)) for large lists.

```python
from typing import List

class Solution:
  def binarySearch(self , arr : List[int] , target : int):
    left = 0
    right = len(arr) - 1

    while left <= right:
      mid_index = (left + right) // 2

      if mid_index < 0 or mid_index >= len(arr):
          return f"Target not found : {-1}"

      middle_element = arr[mid_index]

      if middle_element == target:
        return f"Target = {target} found on index {mid_index}"
      elif middle_element < target:
        left = mid_index + 1
      else:
        right = mid_index - 1

    return f"Target not found : {-1}"

s = Solution()
```

```python
print(s.binarySearch(arr = [2 , 5 , 8 , 12 , 16 , 21 , 27 , 35 , 40] , target = 20))
```
